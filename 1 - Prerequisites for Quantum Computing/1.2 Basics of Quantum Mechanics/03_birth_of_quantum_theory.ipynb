# Chapter 03: The Birth of Quantum Theory

The experiments in the previous notebook created a crisis in physics. Classical physics could not explain blackbody radiation, the photoelectric effect, atomic spectra, or the stability of atoms. The rules that worked perfectly for cannonballs and planets completely failed for atoms and light.

Something fundamentally new was needed. And between 1900 and 1930, a group of physicists invented it. They did not want to. Most of them resisted their own discoveries. But the experiments left them no choice.

This notebook tells the story of how quantum theory was born, idea by idea, puzzle by puzzle. It is one of the greatest stories in the history of human thought.

---

## 1. Max Planck and the Quantum of Energy (1900)

The quantum revolution began with an act of desperation.

### The Problem

Max Planck was trying to derive a formula for blackbody radiation — the spectrum of light emitted by a hot object. The classical formula (Rayleigh-Jeans law) produced the ultraviolet catastrophe: infinite energy at high frequencies. This was clearly wrong.

Planck tried everything he could think of within classical physics. Nothing worked.

### The Desperate Trick

In October 1900, Planck made a mathematical assumption that he did not actually believe was physically true. He assumed that the vibrating atoms in the walls of a blackbody cavity could only emit and absorb energy in discrete chunks — multiples of a fundamental unit.

He wrote: **E = n × h × f**

Where:
- E is the energy
- n is a positive integer (1, 2, 3, ...)
- h is a new constant (now called Planck's constant)
- f is the frequency

The energy can be 1hf, 2hf, 3hf, but never 1.5hf or 2.7hf. Only whole multiples. Like a staircase, not a ramp.

### The Result

With this assumption, Planck derived a formula that perfectly matched the experimental data. The ultraviolet catastrophe disappeared. The formula correctly predicted that high-frequency emission is suppressed.

### Why It Works

Here is the key insight. If energy comes in discrete chunks of size hf, then high-frequency modes require very large energy quanta (because hf is large when f is large). At any given temperature, there is not enough thermal energy available to excite these high-frequency modes. They are frozen out.

Imagine you need to pay for things, but money only comes in fixed denominations. Low-frequency modes are like buying cheap items — easy to afford many of them. High-frequency modes are like buying luxury cars — each one is so expensive that you can afford very few or none at all, even if you have a decent amount of money. The total spending (total energy emitted) stays finite.

### Planck's Reaction

Planck himself did not believe energy quantization was real. He thought it was a mathematical trick that happened to give the right answer. He spent the next decade trying to derive his formula without the quantum hypothesis. He failed.

Planck later wrote that his quantum hypothesis was "an act of desperation" because "a theoretical explanation had to be found at any cost, whatever the price."

### Planck's Constant

The constant h that Planck introduced is one of the most fundamental constants in nature:

**h = 6.626 × 10^-34 joule-seconds**

This number is incredibly tiny. In everyday life, energy quanta are so small that energy appears continuous. It is like saying money comes in units of one-trillionth of a cent — for practical purposes, money is continuous.

But at the atomic scale, where energies are tiny, the quantization becomes obvious and important. The steps on the staircase are big enough relative to the energy scales involved that you cannot ignore them.

The reduced Planck constant, **ℏ = h/(2π)**, is even more common in quantum mechanics formulas. You will see it everywhere.

## 2. Albert Einstein and the Photon (1905)

Planck quantized the emission and absorption of light. Einstein went further. He said that light itself is quantized.

### Einstein's Bold Idea

In 1905, Einstein proposed that light is not a continuous wave. It is made of discrete particles, which he called **quanta of light** (later called photons). Each photon carries energy:

**E = h × f**

A beam of light is not a smooth wave. It is a stream of individual photons. A brighter light has more photons per second. A higher-frequency light has more energetic photons.

### Explaining the Photoelectric Effect

Einstein used the photon idea to explain the photoelectric effect perfectly:

1. A single photon hits a single electron. If the photon's energy (hf) is greater than the electron's binding energy (called the work function), the electron is ejected. If not, nothing happens. This explains the threshold frequency.

2. Brighter light means more photons, which means more electrons are ejected. But each photon still has the same energy, so each electron has the same kinetic energy. This explains why brightness does not affect individual electron energy.

3. The photon hits the electron instantly. No accumulation time needed. This explains the lack of time delay.

The equation is: **KE_max = hf - W**

Where KE_max is the maximum kinetic energy of ejected electrons, hf is the photon energy, and W is the work function (the minimum energy needed to free the electron from the metal).

### The Controversy

Einstein's idea was extremely controversial. Most physicists, including Planck, rejected it. How could light be a particle when interference experiments clearly showed it was a wave? Niels Bohr went so far as to say he would never accept the photon concept.

But the evidence kept mounting. In 1916, Robert Millikan — who actually set out to disprove Einstein — performed meticulous photoelectric effect experiments. His data perfectly confirmed Einstein's equation. The photon was real.

Einstein received the Nobel Prize in 1921, not for relativity, but for his explanation of the photoelectric effect.

### What This Means

Light is not a continuous wave. It is made of particles. But it also interferes like a wave. How can it be both?

This is the heart of quantum mechanics. Light is neither purely a wave nor purely a particle. It is a new kind of thing — a quantum object — that behaves like a wave in some experiments and like a particle in others. We will explore this deeply in notebook 04.

## 3. Niels Bohr and the Quantum Atom (1913)

Planck quantized energy. Einstein quantized light. Bohr quantized the atom.

### The Problem

Rutherford's planetary model of the atom had two fatal flaws:
1. Classical electromagnetism says orbiting electrons should radiate energy and spiral into the nucleus
2. The model could not explain why atoms emit light at only specific frequencies (the discrete spectral lines)

### Bohr's Postulates

In 1913, Niels Bohr proposed a modified atomic model with three radical postulates:

**Postulate 1:** Electrons orbit the nucleus in specific, allowed orbits. While in these orbits, they do not radiate energy. This directly contradicts classical electromagnetism, and Bohr had no explanation for why. He just said: this is how it works.

**Postulate 2:** The allowed orbits are determined by a quantization condition. The angular momentum of the electron must be a whole multiple of ℏ (h/2π):

**L = n × ℏ** where n = 1, 2, 3, ...

The integer n is called the **principal quantum number**. n = 1 is the lowest orbit (closest to the nucleus), n = 2 is the next, and so on.

**Postulate 3:** An electron can jump from one orbit to another by absorbing or emitting a photon. The photon's energy equals the energy difference between the two orbits:

**E_photon = E_n2 - E_n1 = h × f**

### The Energy Levels of Hydrogen

Using these postulates, Bohr calculated the energy levels of hydrogen:

**E_n = -13.6 eV / n²**

The negative sign means the electron is bound to the nucleus. The lowest energy (n=1) is called the ground state. Higher values of n are excited states.

Key values:
- n = 1: E = -13.6 eV (ground state)
- n = 2: E = -3.4 eV
- n = 3: E = -1.51 eV
- n = ∞: E = 0 eV (electron is free, no longer bound)

### Explaining the Spectrum

When an electron jumps from level n=3 to n=2, it emits a photon with energy:

E = -1.51 - (-3.4) = 1.89 eV

This corresponds to a photon with wavelength 656 nm — the red line in hydrogen's spectrum.

Bohr's formula perfectly reproduced the Balmer formula that had been known empirically since 1885. For the first time, there was a theoretical explanation for why atoms emit only certain frequencies.

### The Limitations

Bohr's model worked beautifully for hydrogen (one electron) but failed for larger atoms. It could not explain why only certain orbits are allowed. It mixed classical and quantum ideas in an inconsistent way.

The model was a stepping stone, not the final answer. The full answer would come from the wave mechanics of Schrödinger and the matrix mechanics of Heisenberg in the 1920s.

### The Analogy

Think of the energy levels like the floors in a building. You can be on floor 1, floor 2, or floor 3, but never on floor 1.5. To go from floor 1 to floor 3, you need a specific amount of energy (climbing two floors). When you jump back down, you release that same energy as a photon.

There is no elevator that goes to any continuous height. Only discrete floors exist.

## 4. Louis de Broglie and Matter Waves (1924)

If light, which everyone thought was a wave, turned out to also be a particle (photon), then could things we think are particles also be waves?

Louis de Broglie asked this question in his PhD thesis in 1924. His answer changed physics forever.

### The Hypothesis

De Broglie proposed that every particle has an associated wavelength:

**λ = h / p = h / (m × v)**

Where λ is the wavelength, h is Planck's constant, p is the momentum, m is the mass, and v is the velocity.

This is called the **de Broglie wavelength**.

### What This Means

Every moving object — every electron, every atom, every baseball, every person — has a wave associated with it. The wave is not like a sound wave or a water wave. It is a wave of probability (though that interpretation came later).

### Why We Do Not Notice It

Because Planck's constant h is so incredibly tiny (6.626 × 10^-34), the wavelength of everyday objects is absurdly small.

A baseball (mass 0.145 kg) moving at 40 m/s has a de Broglie wavelength of about 10^-34 meters. For comparison, an atom is about 10^-10 meters. The baseball's wavelength is a trillion trillion times smaller than an atom. You would never notice the wave behavior.

But an electron (mass 9.11 × 10^-31 kg) moving at 10^6 m/s has a wavelength of about 7 × 10^-10 meters. This is comparable to the spacing between atoms in a crystal. At this scale, wave effects (interference, diffraction) become observable.

### Experimental Confirmation

In 1927, Clinton Davisson and Lester Germer fired electrons at a nickel crystal and observed a diffraction pattern — exactly the pattern you would expect from a wave with the wavelength de Broglie predicted.

Electrons, which everyone knew were particles, produced interference patterns. This was conclusive evidence that matter has wave properties.

### Explaining Bohr's Orbits

De Broglie's idea gave a beautiful explanation for Bohr's quantization condition. If the electron is a wave, then the allowed orbits are those where the electron wave forms a standing wave — where a whole number of wavelengths fit exactly around the orbit.

Think of a guitar string. It can only vibrate at certain frequencies because the wave must fit perfectly on the string. A half wavelength, one wavelength, one and a half wavelengths, and so on. Frequencies in between create waves that interfere with themselves and cancel out.

Similarly, an electron wave around a nucleus can only exist if a whole number of wavelengths fit around the orbit. This naturally gives Bohr's condition: L = nℏ.

The quantization of energy is not an arbitrary rule. It is a natural consequence of the electron being a wave.

### Impact

De Broglie's hypothesis completed the wave-particle duality picture:
- Light (thought to be a wave) also has particle properties
- Electrons (thought to be particles) also have wave properties
- Everything in the universe is both wave and particle

This was a profound unification. The wall between particles and waves, which had stood for centuries, was torn down. There is only one kind of stuff in the universe, and it has both properties.

## 5. Werner Heisenberg and Matrix Mechanics (1925)

In 1925, Werner Heisenberg, at age 23, created the first complete mathematical formulation of quantum mechanics.

### The Philosophy

Heisenberg was frustrated with Bohr's model because it talked about electron orbits that nobody could observe. You cannot see an electron orbiting a nucleus. So why base a theory on unobservable things?

Heisenberg decided to build a theory based entirely on observable quantities: the frequencies and intensities of spectral lines that you can actually measure.

### The Approach

Instead of describing the position and velocity of an electron (which are unobservable inside an atom), Heisenberg used matrices to describe the transitions between energy levels. Each observable quantity was represented by a matrix, and the rules for combining these matrices gave predictions that matched experiments.

This was matrix mechanics. It was mathematically rigorous and gave correct answers, but it was extremely abstract and unintuitive. Most physicists found it very difficult to work with.

### The Key Discovery

While working with his matrices, Heisenberg discovered something shocking: the order of multiplication matters.

In ordinary mathematics, 3 × 5 = 5 × 3. Multiplication commutes.

But in Heisenberg's matrix mechanics, position times momentum is NOT the same as momentum times position:

**x × p - p × x = iℏ ≠ 0**

This is the **canonical commutation relation**, and it is the mathematical foundation of the uncertainty principle (which we will cover in notebook 07).

When two quantities do not commute, it means you cannot simultaneously know both of them with perfect precision. The more precisely you know position, the less precisely you know momentum, and vice versa.

This non-commutativity is directly relevant to quantum computing. Quantum gates in general do not commute — applying gate A then gate B gives a different result than applying gate B then gate A.

## 6. Erwin Schrödinger and Wave Mechanics (1926)

Less than a year after Heisenberg published matrix mechanics, Erwin Schrödinger developed an alternative formulation: wave mechanics.

### The Idea

If electrons are waves (as de Broglie proposed), then there should be a wave equation for them, just like there are wave equations for sound waves, water waves, and electromagnetic waves.

Schrödinger found that equation:

**iℏ × (∂Ψ/∂t) = Ĥ × Ψ**

This is the **Schrödinger equation**. It is the master equation of quantum mechanics. It tells you how the wave function Ψ (psi) changes over time.

### What is the Wave Function?

The wave function Ψ(x, t) is a complex-valued function that describes the state of a quantum system. It is defined at every point in space and time.

But what does it mean physically? What is waving? This was a huge puzzle.

Schrödinger initially thought the wave function represented a smeared-out electron — the electron is literally spread across space like a cloud. But this interpretation had problems.

The correct interpretation came from Max Born (covered in the next section).

### The Time-Independent Schrödinger Equation

For systems with fixed energy (called stationary states), the Schrödinger equation simplifies to:

**Ĥ × Ψ = E × Ψ**

This is an eigenvalue equation. The wave function Ψ is the eigenvector (eigenfunction), and E is the eigenvalue (the energy). If you remember eigenvalues from linear algebra, this is exactly the same idea.

When you solve this equation for the hydrogen atom, you naturally get the discrete energy levels that Bohr had postulated. But now they come out of the mathematics automatically, with no ad hoc assumptions.

### Wave Mechanics vs Matrix Mechanics

At first, physicists were confused. Were there two different theories of quantum mechanics? Which one was right?

Schrödinger himself proved in 1926 that the two formulations are mathematically equivalent. They give the same predictions for every experiment. They are just different languages for the same physics, like describing a circle using Cartesian coordinates (x² + y² = r²) or polar coordinates (r = constant). Same circle, different description.

Most physicists preferred Schrödinger's wave mechanics because it was more intuitive and easier to visualize. You can picture a wave function. Matrices are more abstract.

### Schrödinger's Cat (Preview)

Schrödinger himself was deeply uncomfortable with the implications of his own equation. In 1935, he proposed a thought experiment — Schrödinger's cat — to illustrate what he saw as the absurdity of quantum mechanics. We will discuss this in detail in notebook 07.

The irony is rich. The man who invented the most important equation in quantum mechanics spent the rest of his career arguing that quantum mechanics could not possibly be the whole story.

## 7. Max Born and the Probability Interpretation (1926)

The wave function Ψ came from Schrödinger's equation. But what does it mean?

### The Problem

Schrödinger thought Ψ represented the actual distribution of the electron in space — the electron is a smeared-out cloud. But this interpretation failed. When you measure an electron, you always find it at a definite point, not spread out. And a single electron in a diffraction experiment produces a single dot on the screen, not a spread-out pattern.

### Born's Interpretation

In 1926, Max Born proposed a radically different interpretation:

**|Ψ(x)|² is the probability of finding the particle at position x.**

The wave function is not the electron. It is not a physical wave of matter. It is a **probability wave** — a wave of knowledge, of likelihood, of potentiality.

Where |Ψ|² is large, you are likely to find the electron. Where |Ψ|² is small, you are unlikely to find it. Where |Ψ|² is zero, you will never find it.

### This is the Born Rule

This interpretation is called the **Born rule**, and it is one of the most important principles in all of quantum mechanics:

**Probability = |Amplitude|²**

The wave function gives amplitudes (complex numbers). The square of the amplitude gives probability (a real number between 0 and 1).

This is the rule that directly applies to quantum computing. When you measure a qubit in state α|0⟩ + β|1⟩:
- Probability of getting 0 = |α|²
- Probability of getting 1 = |β|²

### The Philosophical Shock

Born's interpretation was deeply disturbing. It meant that quantum mechanics is fundamentally probabilistic. The wave function does not tell you what the electron will do. It only tells you what the electron might do. The outcome of any individual measurement is unpredictable.

This is not probabilistic because of ignorance (like flipping a coin — in principle, you could predict the outcome if you knew all the forces). It is fundamentally probabilistic. Even with perfect, complete knowledge of the wave function, you cannot predict the outcome. Only probabilities.

Albert Einstein was famously unhappy with this. He wrote to Born: "I am convinced that God does not play dice." Born responded: "But it appears that He does."

### Why This Matters

The Born rule is why quantum computing output is probabilistic. When you run a quantum algorithm, you do not get a definite answer. You get a probabilistic answer. To get reliable results, you run the algorithm many times and look at the statistics.

Quantum algorithms work by manipulating the amplitudes so that the correct answer has high probability and the wrong answers have low probability. This manipulation uses interference — constructive interference amplifies the right answer, destructive interference suppresses the wrong ones.

## 8. Paul Dirac and the Unification (1930)

In 1930, Paul Dirac published "The Principles of Quantum Mechanics," which unified Heisenberg's matrix mechanics and Schrödinger's wave mechanics into a single, more abstract framework.

### Dirac's Approach

Dirac introduced the idea that quantum states live in an abstract vector space called a **Hilbert space**. The state of a system is represented by a vector in this space, which he wrote using the now-famous **bra-ket notation**:

- **Ket**: |ψ⟩ represents a quantum state (a column vector)
- **Bra**: ⟨ψ| represents the dual of that state (a row vector, complex conjugated)
- **Bracket**: ⟨φ|ψ⟩ is the inner product (the overlap between two states)

This notation is elegant and powerful. You already learned it in the linear algebra section. Now you know where it comes from.

### Physical Quantities as Operators

Dirac formalized the idea that physical quantities (like energy, momentum, position, and spin) are represented by mathematical operators (matrices in finite dimensions). Measuring a quantity corresponds to applying the operator to the state.

The possible measurement outcomes are the eigenvalues of the operator. The states after measurement are the eigenvectors. This is the formal connection between the linear algebra you learned and quantum physics.

### The Dirac Equation

Dirac also combined quantum mechanics with special relativity to create the Dirac equation. This equation predicted the existence of antimatter (the positron) before it was discovered experimentally. It was one of the greatest theoretical predictions in physics history.

For quantum computing, the non-relativistic Schrödinger equation is sufficient. But it is worth knowing that the theory you are learning is part of a larger, deeper framework.

### Impact on Quantum Computing

Dirac's framework is exactly the framework used in quantum computing:
- Quantum states are vectors in Hilbert space → qubits are vectors in C²
- Physical quantities are operators (matrices) → quantum gates are unitary matrices
- Measurement outcomes are eigenvalues → qubit measurement gives eigenvalues of the measurement operator
- Bra-ket notation → used universally in quantum computing

When you write |0⟩ and |1⟩ in quantum computing, you are using Dirac's notation to describe states in a 2-dimensional Hilbert space. The entire mathematical framework of quantum computing is Dirac's framework applied to two-level systems.

## 9. The Timeline of the Quantum Revolution

Let us put the whole story in order.

| Year | Scientist | Contribution |
|------|-----------|-------------|
| 1900 | Max Planck | Quantized energy in blackbody radiation. Introduced h. |
| 1905 | Albert Einstein | Light is made of photons (E = hf). Explained photoelectric effect. |
| 1911 | Ernest Rutherford | Discovered the atomic nucleus. |
| 1913 | Niels Bohr | Quantized electron orbits in atoms. Explained hydrogen spectrum. |
| 1922 | Stern & Gerlach | Showed angular momentum is quantized. |
| 1923 | Arthur Compton | Showed photons carry momentum. |
| 1924 | Louis de Broglie | All matter has wave properties (λ = h/p). |
| 1925 | Werner Heisenberg | Created matrix mechanics. |
| 1926 | Erwin Schrödinger | Created wave mechanics (Schrödinger equation). |
| 1926 | Max Born | Probability interpretation (|Ψ|² = probability). |
| 1927 | Werner Heisenberg | Uncertainty principle. |
| 1927 | Davisson & Germer | Experimentally confirmed electron diffraction (matter waves). |
| 1928 | Paul Dirac | Relativistic quantum mechanics. Predicted antimatter. |
| 1930 | Paul Dirac | Unified framework. Bra-ket notation. |
| 1935 | Einstein, Podolsky, Rosen | EPR paradox — challenged completeness of QM. |
| 1935 | Schrödinger | Schrödinger's cat thought experiment. Coined "entanglement." |

In just 30 years, physics went from a complete, deterministic, classical worldview to a fundamentally new understanding of reality. The quantum revolution is arguably the greatest intellectual achievement of the 20th century.

### The People

Notice something about the scientists involved. Most of them were young. Heisenberg was 23 when he invented matrix mechanics. Dirac was 25 when he published his equation. Einstein was 26 when he explained the photoelectric effect.

And most of them disagreed with each other. Einstein never accepted the probability interpretation. Schrödinger never accepted wave function collapse. Bohr and Einstein debated quantum mechanics for decades. The theory was built by people who argued constantly.

This is how science works. The truth does not care if the scientists are comfortable with it.

## 10. The Key Ideas to Carry Forward

From the birth of quantum theory, these are the ideas that matter most for quantum computing:

### 1. Quantization

Energy, angular momentum, and other physical quantities come in discrete units. This is why qubits have discrete energy levels (|0⟩ and |1⟩) rather than a continuous range of values.

### 2. Wave-Particle Duality

Everything is both wave and particle. The wave nature allows interference, which is the core mechanism of quantum algorithms. Without wave behavior, quantum computing would not work.

### 3. The Born Rule

Probability = |amplitude|². This is how we extract information from quantum computers. Measurement gives random outcomes, with probabilities determined by the amplitudes.

### 4. The Schrödinger Equation

Quantum states evolve according to a linear equation. Linearity means superposition is possible. If |ψ₁⟩ and |ψ₂⟩ are valid states, then any combination α|ψ₁⟩ + β|ψ₂⟩ is also valid.

### 5. Operators and Eigenvalues

Physical quantities are represented by operators (matrices). Measurement outcomes are eigenvalues. This is the formal connection between linear algebra and quantum physics.

### 6. Non-Commutativity

The order of operations matters. Position and momentum do not commute. This leads to the uncertainty principle and means that the order of quantum gates matters.

### 7. Dirac's Framework

Quantum states are vectors in Hilbert space. This abstract framework is exactly what quantum computing uses.

## Summary

The quantum revolution was built step by step:

- **Planck (1900)**: Energy is quantized in units of hf. Not continuous.
- **Einstein (1905)**: Light is made of particles (photons) with energy E = hf.
- **Bohr (1913)**: Electrons in atoms have quantized energy levels. Transitions emit photons.
- **de Broglie (1924)**: All matter has wave properties. Wavelength = h/p.
- **Heisenberg (1925)**: Matrix mechanics. Non-commuting operators. Observables only.
- **Schrödinger (1926)**: Wave equation for quantum systems. Wave function Ψ.
- **Born (1926)**: |Ψ|² gives probability. Quantum mechanics is fundamentally probabilistic.
- **Dirac (1930)**: Unified framework using Hilbert space and bra-ket notation.

These ideas form the foundation of quantum mechanics and quantum computing.
