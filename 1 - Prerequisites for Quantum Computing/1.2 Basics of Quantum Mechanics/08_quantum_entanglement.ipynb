# Chapter 08: Quantum Entanglement

Schrödinger called it "the characteristic trait of quantum mechanics, the one that enforces its entire departure from classical lines of thought."

Einstein called it "spooky action at a distance."

We call it **entanglement**.

It is the phenomenon where two or more quantum particles become linked in such a way that the quantum state of each particle cannot be described independently of the others. They share a single, unified quantum state, no matter how far apart they are.

This is the fuel that powers the exponential speedup of quantum computing.

---

## 1. What is Entanglement?

### A Classical Analogy (That Fails)

Imagine I have a pair of gloves, one left and one right. I put them in separate boxes and mail one to Alice in New York and one to Bob in Tokyo.

Alice opens her box and sees a **Left** glove. She immediately knows that Bob has the **Right** glove.

Is this spooky? No. The gloves were determined from the start. One was always left, one was always right. We just didn't know which was which.

### The Quantum Version

Now imagine I have two quantum coins (qubits) in an entangled state. I send one to Alice and one to Bob.

Alice measures her coin. It is **Heads**. She immediately knows Bob's coin is **Tails**.

The difference? **The coins were not Heads or Tails before the measurement.** They were in a superposition of both.

Wait, you say. Maybe they were Heads/Tails all along, and we just didn't know (like the gloves)?

John Bell proved in 1964 that this "hidden variable" explanation is mathematically impossible. The coins literally do not have a fixed state until measured. And yet, their outcomes are perfectly correlated.

When Alice measures her coin, she forces it to choose a state. Instantly, Bob's coin chooses the opposite state. This happens instantaneously, even if they are light-years apart.

## 2. The Mathematics of Entanglement

### Product States (Not Entangled)

If we have two qubits, their combined state is usually written as a tensor product.

If qubit 1 is in state |ψ⟩ and qubit 2 is in state |φ⟩, the combined state is:

|Ψ⟩ = |ψ⟩ ⊗ |φ⟩

Example: |0⟩ ⊗ |0⟩ = |00⟩
Example: ((|0⟩ + |1⟩)/√2) ⊗ ((|0⟩ + |1⟩)/√2) = 0.5(|00⟩ + |01⟩ + |10⟩ + |11⟩)

These are called **product states**. The measurements on qubit 1 are independent of qubit 2.

### Entangled States

An entangled state is a state that **cannot** be written as a product of two individual states.

The most famous example is the **Bell State** (or EPR pair):

|Φ⁺⟩ = (|00⟩ + |11⟩) / √2

Try to factor this into (a|0⟩ + b|1⟩) ⊗ (c|0⟩ + d|1⟩). You can't. It's impossible.

This means neither qubit has a state of its own. Only the pair has a state.

Let's visualize the setup of an entanglement experiment.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_entanglement_setup():
    fig, ax = plt.subplots(figsize=(10, 4))
    
    # Source
    ax.add_patch(patches.Circle((5, 2), 0.5, color='gold', zorder=10))
    ax.text(5, 2, 'Source', ha='center', va='center', fontweight='bold')
    
    # Alice
    ax.add_patch(patches.Rectangle((1, 1.5), 1, 1, facecolor='lightblue', edgecolor='black'))
    ax.text(1.5, 2, 'Alice', ha='center', va='center', fontweight='bold')
    
    # Bob
    ax.add_patch(patches.Rectangle((8, 1.5), 1, 1, facecolor='lightgreen', edgecolor='black'))
    ax.text(8.5, 2, 'Bob', ha='center', va='center', fontweight='bold')
    
    # Entangled Particles
    # Left particle
    ax.arrow(4.5, 2, -1.8, 0, head_width=0.1, head_length=0.2, fc='k', ec='k')
    ax.text(3.5, 2.2, 'Particle A', ha='center')
    
    # Right particle
    ax.arrow(5.5, 2, 1.8, 0, head_width=0.1, head_length=0.2, fc='k', ec='k')
    ax.text(6.5, 2.2, 'Particle B', ha='center')
    
    # Wave connection
    x = [2.2, 3, 4, 5, 6, 7, 7.8]
    y = [2, 1.8, 2.2, 1.8, 2.2, 1.8, 2]
    # Just an artistic squiggly line to imply connection
    # Not plotting it mathematically to keep it simple diagram
    
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 4)
    ax.axis('off')
    plt.title("The Bell Test Setup: Spooky Action at a Distance")
    plt.show()

draw_entanglement_setup()
```

### What does the math say?

|Φ⁺⟩ = (|00⟩ + |11⟩) / √2

This superposition tells us:
1. If you measure, you get |00⟩ with 50% probability (Alice=0, Bob=0).
2. Or you get |11⟩ with 50% probability (Alice=1, Bob=1).
3. You NEVER get |01⟩ or |10⟩.

Alice's result is random (50/50). Bob's result is random (50/50). But their results are perfectly correlated.

If Alice measures 0, Bob MUST measure 0.
If Alice measures 1, Bob MUST measure 1.

```python
import numpy as np

def simulate_bell_state_measurements(n_trials=10):
    print(f"Simulating {n_trials} measurements of Bell State |Φ+> = (|00> + |11>) / sqrt(2)\n")
    
    print("Trial | Alice | Bob  | Correlation")
    print("------|-------|------|------------")
    
    for i in range(n_trials):
        # The universe flips a coin
        outcome = np.random.choice(['00', '11'])
        
        alice_result = outcome[0]
        bob_result = outcome[1]
        
        match = "Perfect" if alice_result == bob_result else "FAIL"
        
        print(f"{i+1:5} |   {alice_result}   |   {bob_result}  | {match}")

simulate_bell_state_measurements()
```

## 3. Einstein, Podolsky, Rosen (EPR) Paradox

In 1935, Einstein, Podolsky, and Rosen (EPR) wrote a paper arguing that quantum mechanics must be incomplete.

Their argument:
1. If Alice measures her qubit, she knows Bob's qubit state instantly.
2. Information cannot travel faster than light.
3. Therefore, Bob's qubit must have already been in that state before Alice measured.
4. Quantum mechanics says it wasn't. It says the state was undefined.
5. Therefore, quantum mechanics is missing something. There must be "hidden variables" that determine the outcome in advance.

They concluded that "God does not play dice" and that the particles are like the gloves in the boxes.

For 30 years, this was just a philosophical debate. Then John Bell changed everything.

---

## 4. Bell's Theorem: Proving Einstein Wrong

In 1964, John Bell derived a mathematical inequality (Bell's Inequality) that MUST be satisfied if the "hidden variable" (gloves) theory is true.

Basically, if objects have definite properties before you measure them, there is a limit to how correlated they can be when measured in different ways (different angles).

Quantum mechanics predicts correlations that violate this inequality. It predicts stronger correlations than any classical hidden variable theory allows.

### The Experiments

In the 1980s (Alain Aspect) and 2010s, experiments confirmed it: **The universe violates Bell's inequality.**

This means:
1. There are no hidden variables (the coins are not pre-determined).
2. Or, the measurement affects the other particle instantly (non-locality).

Most physicists accept **non-local realism**: particles do not have defined properties until measured, and they are connected non-locally. Einstein was wrong about the "hidden variables," but right that it is spooky.

---

## 5. Can We Use This to Communicate Faster Than Light?

If Alice's measurement instantly affects Bob's qubit, can she send a message? 

**No.** This is the "No-Communication Theorem."

Why?
When Alice measures, she gets a RANDOM result (0 or 1). She cannot force her qubit to be 0 or 1. She just sees what she gets.

Bob also gets a RANDOM result. He sees a random stream of 0s and 1s. He has no way of knowing if his 0 is because Alice measured 0, or because he just happened to get 0 randomly.

They only see the correlation when they meet up later and compare their notebooks. "Hey, look, every time I got 1, you got 1!"

So nature allows "spooky action," but forbids using it to send instantaneous text messages. Relativity is safe.

---

## 6. Quantum Teleportation

While we can't communicate FTL, we CAN use entanglement to standard teleportation. Not of matter (sorry, Star Trek), but of **information**.

**Quantum Teleportation** allows Alice to send the *state* of a qubit to Bob, using an entangled pair and two bits of classical information.

**The Recipe:**
1. Alice and Bob share an entangled pair (A and B).
2. Alice has a mystery qubit Q she wants to send.
3. Alice performs a special measurement on Q and A.
4. She gets two classical bits (00, 01, 10, or 11).
5. She calls Bob (classical channel) and tells him the two bits.
6. Bob applies a correction to B depending on the bits.
7. Bob's qubit B transforms into the exact state of Q.

Note: The original qubit Q is destroyed (no-cloning). The information moves from Alice to Bob. It requires classical communication, so it is not faster than light.

---

## 7. Summary

- **Entanglement** is a strong correlation between quantum particles where their individual states are undefined.
- **Bell States** (like (|00⟩+|11⟩)/√2) are maximally entangled.
- **Bell's Theorem** proves that these correlations cannot be explained by classical "hidden variables" (pre-determined properties).
- **Non-locality:** Measurement of one particle instantly determines the state of the other, regardless of distance.
- **No FTL Communication:** You cannot use this to send messages faster than light because the outcomes are random.
- **Application:** Entanglement is the resource for quantum teleportation, superdense coding, and the massive state space of quantum computers.
