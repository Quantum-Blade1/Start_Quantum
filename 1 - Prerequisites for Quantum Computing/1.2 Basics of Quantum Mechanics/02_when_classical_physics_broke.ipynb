# Chapter 02: When Classical Physics Broke

By the late 1800s, classical physics seemed almost perfect. Physicists had successfully explained mechanics, thermodynamics, electromagnetism, and optics. Many believed that the major work of physics was done. All that remained was to refine the measurements and add more decimal places.

They were completely wrong.

In the span of about 30 years, a series of experiments produced results that classical physics simply could not explain. These were not minor disagreements or small errors. These were fundamental failures. The predictions of classical physics were not just slightly off. They were absurdly, catastrophically wrong.

These failures forced physicists to accept that the universe does not work the way they thought. The rules that work so well for planets, cars, and bridges completely fall apart when you look at atoms, electrons, and light.

This notebook covers the experiments that broke classical physics. Each one is a crack in the foundation. Together, they brought the whole building down.

## 1. The Ultraviolet Catastrophe: Infinite Energy from a Hot Object

This is the experiment that directly led to the invention of quantum mechanics. It is the most important failure in the history of physics.

### The Problem: Blackbody Radiation

Every object that has a temperature emits electromagnetic radiation (light, infrared, etc.). When you heat a piece of iron, it first glows dull red, then orange, then yellow, then white as it gets hotter. The sun emits mostly visible light because of its surface temperature. Your body emits infrared radiation, which is why night-vision cameras can see you in the dark.

A **blackbody** is an idealized object that absorbs all radiation that hits it and emits radiation based purely on its temperature. It is a perfect absorber and a perfect emitter. A good approximation of a blackbody is a small hole in a heated cavity — the hole absorbs almost all light that enters, and the light that comes out depends only on the temperature.

Scientists in the late 1800s carefully measured the spectrum of blackbody radiation — how much energy is emitted at each frequency. They found a smooth, bell-shaped curve. At low frequencies, not much energy is emitted. At intermediate frequencies, the emission peaks. At very high frequencies, the emission drops back down to nearly zero.

### What Classical Physics Predicted

Classical physics (specifically, the Rayleigh-Jeans law) predicted that the energy emitted at a given frequency should increase as the frequency increases. Higher frequency means more energy. And since there is no upper limit to frequency, the energy should keep going up forever.

This means that according to classical physics, any hot object should emit infinite total energy, with most of it in the ultraviolet and beyond. A warm cup of coffee should be emitting deadly gamma rays. A candle should outshine the sun.

This is obviously absurd. And it was called the **ultraviolet catastrophe** because the prediction diverges (goes to infinity) at ultraviolet frequencies.

### Why Classical Physics Gets This Wrong

The problem is the assumption that energy is continuous. In classical physics, each mode of vibration in the cavity can hold any amount of energy. Since there are infinitely many high-frequency modes, they collectively absorb and emit infinite energy.

Imagine it this way. Suppose you are at a buffet, and every dish can hold any amount of food. There are infinitely many small dishes (high-frequency modes). Even if each one holds a tiny amount, infinitely many tiny amounts add up to infinity.

### The Real World

In reality, a hot object does NOT emit infinite energy. The emission peaks at a certain frequency and then drops off sharply at higher frequencies. The high-frequency modes are somehow suppressed.

Classical physics had no explanation for this. Something fundamentally new was needed.

Let's visualize this catastrophe in Python.

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
h = 6.626e-34
c = 3e8
k = 1.38e-23
T = 5000 # Temperature in Kelvin (Sun-like)

wavelengths = np.linspace(1e-9, 3e-6, 1000) # 1nm to 3000nm

# Rayleigh-Jeans Law (Classical Prediction)
# Intensity ~ T / lambda^4
def rayleigh_jeans(lam, T):
    return (2 * c * k * T) / lam**4

# Planck's Law (Quantum Reality)
# Intensity ~ 1 / (lambda^5 * (exp(hc/lkT) - 1))
def planck(lam, T):
    a = 2.0 * h * c**2
    b = (h * c) / (lam * k * T)
    return a / ( (lam**5) * (np.exp(b) - 1.0) )

RJ = rayleigh_jeans(wavelengths, T)
PL = planck(wavelengths, T)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(wavelengths*1e9, PL, label='Planck Law (Reality)', color='green', linewidth=3)
plt.plot(wavelengths*1e9, RJ, label='Rayleigh-Jeans (Classical)', color='red', linestyle='--', linewidth=2)

# Limit y-axis because Classical goes to infinity
plt.ylim(0, np.max(PL)*1.5)
plt.title(f'The Ultraviolet Catastrophe (T={T}K)')
plt.xlabel('Wavelength (nm)')
plt.ylabel('Spectral Radiance')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

**Visualization:** The Red dashed line is what classical physics predicted: as wavelength gets smaller (UV), energy shoots to infinity. This is the catastrophe.
The Green line is what actually happens (Planck's Law). Quantum mechanics saves the universe from infinite energy.

## 2. The Photoelectric Effect: Light Behaves Like Particles

The photoelectric effect was explained by Albert Einstein in 1905. It is the experiment that proved light behaves like particles, even though Maxwell's theory had convincingly shown that light is a wave.

### The Experiment

When you shine light on a metal surface, electrons are ejected from the surface. This is the photoelectric effect. The ejected electrons can be collected and measured.

Simple enough. But the details were deeply puzzling.

### What Classical Physics Predicted

Light is a wave, and the energy of a wave depends on its amplitude (brightness). So:

1. Brighter light should eject electrons with more energy (bigger wave = more energy = faster electrons)
2. Any frequency of light should work if you make it bright enough (enough energy from any color)
3. There should be a time delay — dim light needs time to accumulate enough energy to eject an electron

### What Actually Happens

Every single prediction was wrong.

1. **Brighter light does NOT eject faster electrons.** More intense light ejects MORE electrons, but each individual electron has the SAME energy regardless of brightness. Making the light brighter is like sending more bullets, not bigger bullets.

2. **Below a certain frequency, NO electrons are ejected, no matter how bright the light.** You can shine the brightest red light in the universe on certain metals and nothing happens. But even a tiny amount of ultraviolet light immediately ejects electrons. The frequency (color) matters. The brightness does not (for whether ejection happens at all).

3. **There is NO time delay.** Even with incredibly dim light, electrons are ejected instantaneously (within nanoseconds). There is no build-up period.

### Why This Breaks Classical Physics

If light were purely a wave, its energy would depend on amplitude (brightness). But the photoelectric effect shows that the energy delivered to each electron depends on frequency (color), not amplitude. This makes no sense for a wave.

The threshold frequency is especially damning. In the wave picture, any frequency should work if the light is bright enough — you just need to accumulate enough wave energy. But nature says no. Below the threshold frequency, no amount of brightness will work. Ever.

The instantaneous ejection is also impossible for a wave. A dim wave should need time to build up enough energy to eject an electron. But electrons come out immediately.

### Einstein's Explanation

Einstein proposed that light is not a continuous wave. It comes in discrete packets of energy called **photons**. Each photon carries an energy of:

**E = h × f**

Where h is Planck's constant and f is the frequency.

A photon hits an electron, and if the photon has enough energy (high enough frequency), it kicks the electron out. If the photon does not have enough energy (too low frequency), the electron stays. Making the light brighter sends more photons but does not change the energy of each individual photon.

This is like throwing tennis balls at a window. The frequency is like the speed of each ball. The brightness is like the number of balls per second. If each ball is too slow, no single ball will break the window, no matter how many you throw. You need each individual ball to be fast enough.

### The Confusion

This created a serious problem. Maxwell's theory said light is a wave. Young's double-slit experiment (which we will discuss in notebook 04) proved light is a wave through interference. But now the photoelectric effect says light is made of particles.

Which is it? Is light a wave or a particle?

The answer, as we will see, is: both. And neither. It is something new that does not fit into either classical category. But that is for later notebooks.

## 3. Atomic Spectra: Electrons Can Only Have Certain Energies

When you heat a gas (like hydrogen or neon) until it glows, the light it emits is not a continuous rainbow. Instead, it contains only specific colors — specific frequencies. Each element has its own unique set of spectral lines, like a fingerprint.

### What You See

If you look at the light from a hydrogen lamp through a prism, you do not see a smooth spectrum of all colors. You see specific, sharp lines of color:

- A red line at 656 nm
- A blue-green line at 486 nm
- A violet line at 434 nm
- A deeper violet line at 410 nm

And that is it. No other colors. Just these specific lines.

### What Classical Physics Says

If an electron orbiting the nucleus can have any energy (continuous energy), and if it emits light by losing energy, then it should be able to emit light at any frequency. The spectrum should be continuous — all colors, all frequencies, a smooth rainbow.

But hydrogen does not emit a smooth rainbow. It emits only those specific lines.

### What This Means

The electrons in an atom can only have specific, discrete energies. Not just any energy, but only certain allowed values. When an electron jumps from a higher allowed energy to a lower one, it emits a photon with exactly the energy difference between those two levels.

Since the photon's energy determines its frequency (E = hf), and only certain energy differences are allowed, only certain frequencies of light are emitted. Hence the discrete spectral lines.

Imagine a staircase versus a ramp. On a ramp, you can stand at any height. On a staircase, you can only stand at specific heights. Atomic energy levels are like a staircase. You can be on step 1, step 2, or step 3, but never between them.

### The Balmer Formula

In 1885, Johann Balmer found a mathematical formula that predicted the visible spectral lines of hydrogen:

**1/λ = R × (1/2² - 1/n²)** where n = 3, 4, 5, ...

R is the Rydberg constant. This formula worked perfectly, but nobody knew WHY. It was a pattern with no explanation.

The explanation came from Bohr's model of the atom in 1913 (covered in notebook 03). But the key point here is: the discrete spectrum of atoms is a direct message from nature saying that energy inside atoms is not continuous. It is quantized. It comes in specific, fixed amounts.

### Why This Is Important for Quantum Computing

Discrete energy levels are exactly what make qubits possible. A qubit is typically made from a quantum system that has two distinct energy levels. The electron is either in the lower level (which we call |0>) or the upper level (which we call |1>). If energy were continuous, we would not have these clean, distinct states to work with.

## 4. The Stability of Atoms: Why Do We Exist?

We touched on this in the previous notebook, but it deserves its own section because it is such a fundamental problem.

### The Problem

Rutherford showed that atoms have a tiny, dense, positively charged nucleus with electrons orbiting around it. But classical electromagnetism says that any charged particle that accelerates must radiate electromagnetic energy.

An orbiting electron is constantly accelerating (it is changing direction). So it should constantly radiate energy. As it loses energy, it should spiral inward and crash into the nucleus.

Calculations show this should happen in about 10^-11 seconds. That is one hundred-billionth of a second.

### The Consequence

If classical physics were correct, atoms could not exist for more than a tiny fraction of a second. Matter would collapse. Stars, planets, people, everything would cease to exist.

The fact that you are reading this sentence is proof that classical physics is wrong.

### Why This Matters

This is not a minor technical problem. This is classical physics failing to explain the most basic fact of our existence: that matter is stable. That atoms hold together. That chemistry works. That we are here.

The solution is quantum mechanics. Specifically, the idea that electrons do not have definite orbits. They exist in probability clouds called orbitals. The uncertainty principle prevents them from collapsing into the nucleus (more on this in notebook 07). And the quantization of energy means they can only have specific allowed energies, not any energy they want.

Classical physics says atoms should collapse. Quantum mechanics explains why they do not. Without quantum mechanics, stable matter is impossible.

## 5. Specific Heat of Solids: The Frozen-Out Degrees of Freedom

This is a less famous failure, but it is historically important because it was one of the early hints that classical physics was wrong.

### What is Specific Heat?

Specific heat is the amount of energy you need to add to a substance to raise its temperature by one degree. It tells you how much a material resists heating up.

Water has a high specific heat — it takes a lot of energy to heat up. That is why oceans stabilize climate. Metal has a low specific heat — it heats up quickly. That is why a metal handle gets hot fast in the sun.

### The Classical Prediction

Classical statistical mechanics (the Dulong-Petit law) predicts that the specific heat of a solid should be the same for all elements and should not depend on temperature. The value should be about 25 joules per mole per degree.

At room temperature, this prediction works well for most metals.

### The Problem

At low temperatures, the specific heat drops dramatically. For diamond, the specific heat at room temperature is already much lower than the classical prediction. And as temperature approaches absolute zero, the specific heat of all solids approaches zero.

Classical physics says specific heat should be constant. Nature says it is not.

### Why Classical Physics Fails

In classical physics, every mode of vibration of the atoms in a solid contributes equally to the specific heat. But if the vibration energies are quantized (come in discrete chunks), then at low temperatures, there is not enough thermal energy available to excite the high-frequency vibrations. These vibrations are "frozen out" — they cannot participate because the energy quanta are too large.

Imagine a staircase where each step is 1 meter tall. If you only have the energy to climb 50 centimeters, you cannot go up at all. You are frozen at the bottom. The step is too big for the energy you have.

Einstein solved this problem in 1907 by applying Planck's quantum hypothesis to the vibrations of atoms in a solid. By assuming the vibration energies are quantized, he correctly predicted that specific heat decreases at low temperatures.

This was one of the first applications of quantum ideas beyond radiation, showing that quantization is not just about light. It is about everything.

## 6. The Compton Effect: Light Carries Momentum Like a Particle

In 1923, Arthur Compton performed an experiment that provided even stronger evidence that light behaves like a particle.

### The Experiment

Compton fired X-rays at electrons. When an X-ray photon hits an electron, something happens that is impossible to explain if light is purely a wave.

After the collision:
1. The X-ray photon bounces off the electron at a different angle
2. The photon's wavelength increases (its frequency decreases, meaning it loses energy)
3. The electron recoils, gaining energy and momentum

### Why This Is a Problem for Classical Waves

This collision behaves exactly like two billiard balls hitting each other. The photon acts like a particle with definite momentum. It bounces off the electron, transferring some of its energy and momentum.

In the wave picture, light should scatter without changing frequency. A wave hits an object and bounces back at the same frequency (like an echo at the same pitch). But Compton observed a clear change in frequency.

### The Explanation

Compton treated the photon as a particle with energy E = hf and momentum p = h/λ. He used the same conservation laws (energy and momentum) that you would use for billiard ball collisions. The equations perfectly predicted the observed change in wavelength.

This confirmed that photons are not just energy packets. They also carry momentum. They are genuine particles in every meaningful sense — except that they also behave like waves in other experiments.

### The Growing Paradox

By 1923, the evidence was clear:
- Interference and diffraction prove light is a wave
- The photoelectric effect proves light is a particle
- The Compton effect proves light is a particle with momentum

Light is both a wave and a particle. This contradiction had no resolution within classical physics.

## 7. Radioactivity and Nuclear Decay: Fundamental Randomness

In the late 1800s and early 1900s, scientists discovered that some atoms spontaneously emit particles and radiation. This is radioactivity.

### What Happens

An unstable atomic nucleus can spontaneously emit:
- Alpha particles (two protons + two neutrons)
- Beta particles (electrons or positrons)
- Gamma rays (high-energy photons)

When a nucleus decays, it transforms into a different element. Uranium becomes thorium. Carbon-14 becomes nitrogen-14.

### The Classical Problem: When Will an Atom Decay?

Here is the deeply strange thing about radioactive decay: it is absolutely, fundamentally random.

You cannot predict when a specific atom will decay. You can only say the probability. A carbon-14 atom has a half-life of about 5,730 years. This means that after 5,730 years, there is a 50% chance it has decayed. But you cannot say whether it will decay tomorrow or in 20,000 years.

And this is not because we lack information. It is not that some hidden variable determines when the atom will decay and we just do not know what it is. As far as quantum mechanics tells us, the decay is genuinely, fundamentally random. God plays dice, despite what Einstein said.

### Why This Breaks Classical Physics

In classical physics, the universe is deterministic. If you know all the forces and positions, you can predict everything. Randomness is just ignorance — if we knew more, we could predict better.

But radioactive decay says: no. Even with perfect, complete knowledge of an atom, you cannot predict when it will decay. The randomness is built into the fabric of reality. It is not ignorance. It is the way the universe works.

This fundamental randomness is one of the most disturbing and important features of quantum mechanics. And it is directly relevant to quantum computing, where measurement outcomes are inherently probabilistic.

### Analogy

Imagine you have a box of popcorn kernels. You heat them up. Each kernel will pop at some random time. You cannot predict exactly when any specific kernel will pop, but you can predict what fraction of kernels will have popped after a given time.

Radioactive decay is like this, except the randomness is not because the kernels are complex systems we do not fully understand. Each kernel is identical, in an identical environment, and the popping is still random. There is no hidden reason. It just is random.

## 8. Stern-Gerlach Experiment: Quantized Angular Momentum

This experiment, performed in 1922, showed something deeply strange about the properties of atoms.

### The Setup

Otto Stern and Walther Gerlach sent a beam of silver atoms through a non-uniform magnetic field. A non-uniform magnetic field exerts a force on a magnetic dipole (like a tiny bar magnet) that depends on the orientation of the dipole.

Since each silver atom has a magnetic moment (due to its unpaired electron), the atoms should be deflected by the magnetic field.

### What Classical Physics Predicts

If the atoms are randomly oriented (like tiny bar magnets pointing in random directions), the beam should spread out into a continuous smudge on the detector. Some atoms pointing up get deflected up. Some pointing down get deflected down. Most are at some angle in between. The result should be a smooth, continuous band.

### What Actually Happened

The beam split into exactly TWO discrete spots. Not a smooth band. Two spots. Up and down. Nothing in between.

### What This Means

The angular momentum (spin) of the electron is not continuous. It does not take any value like a classical spinning top. It is quantized into exactly two values: spin-up and spin-down.

This is not like a compass needle that can point in any direction. It is like a coin that can only be heads or tails. There is no in-between. No 45 degrees. No "slightly tilted." Just two values.

### Why This Is Huge

This experiment directly shows that quantum properties are quantized — they take only specific, discrete values. And specifically, it shows that spin (a quantum property with no classical analog) has only two values for a spin-1/2 particle.

This is exactly what makes a qubit possible. A qubit is a quantum two-level system. Spin-up = |0>, spin-down = |1>. The Stern-Gerlach experiment was, in a sense, the first observation of a qubit.

### Connection to Measurement

There is another deep lesson here. If you first separate atoms into spin-up and spin-down using a vertical magnetic field, and then take the spin-up atoms and send them through a horizontal magnetic field, they split again — 50/50 into spin-left and spin-right.

But here is the strange part. If you then take the spin-right atoms and send them through a vertical field again, they split 50/50 into spin-up and spin-down. The horizontal measurement somehow erased the information about the original vertical measurement.

In classical physics, measuring a property does not disturb other properties. But in quantum mechanics, measuring one thing (horizontal spin) destroys information about a complementary thing (vertical spin). This is the uncertainty principle in action, and we will explore it deeply in notebook 07.

## 9. The Pattern of Failures

Let us step back and see the pattern in all these failures.

### What Went Wrong

Every single failure has a common theme: **continuity breaks down at small scales.**

| Experiment | Classical Prediction | Reality |
|-----------|---------------------|--------|
| Blackbody radiation | Infinite energy at high frequencies | Energy is finite, peaks and drops off |
| Photoelectric effect | Any frequency works if bright enough | Below threshold frequency, nothing works |
| Atomic spectra | Continuous rainbow | Discrete spectral lines |
| Atom stability | Atoms collapse in 10^-11 seconds | Atoms are stable |
| Specific heat | Constant at all temperatures | Drops to zero at low temperature |
| Compton scattering | Wave scatters without changing frequency | Photon bounces like a particle |
| Radioactive decay | Deterministic and predictable | Fundamentally random |
| Stern-Gerlach | Continuous spread of orientations | Exactly two discrete values |

### The Root Cause

Classical physics assumes:
1. Energy is continuous
2. Properties are always definite
3. The future is determined by the present
4. Measurement does not change the system
5. Light is a wave, electrons are particles, and that is that

Every single one of these assumptions is wrong.

### The New Rules (Preview)

The experiments demand:
1. Energy comes in discrete chunks (quanta)
2. Properties are indefinite until measured
3. The future is fundamentally probabilistic
4. Measurement changes the system
5. Everything is both wave and particle

These are the rules of quantum mechanics. The next notebook covers how physicists figured them out.

## 10. Why Should a Quantum Computing Student Care About This?

You might be thinking: "I want to learn quantum computing, not physics history. Why do I need to know about the ultraviolet catastrophe or the photoelectric effect?"

Here is why.

### Understanding Why Qubits Work

A qubit is a quantum two-level system. It exists because energy is quantized (discrete energy levels). If energy were continuous, there would be no clean two-level systems, and qubits would not exist.

### Understanding Why Measurement is Probabilistic

When you measure a qubit, the result is random. Not because you lack information. Not because your equipment is imperfect. Fundamentally random. This is the lesson from radioactive decay and the Born rule. If you do not understand that quantum randomness is not the same as classical randomness, you will misunderstand quantum algorithms.

### Understanding Why Quantum Computers Are Different

Classical computers are built on classical physics: deterministic, continuous, measurable without disturbance. Quantum computers are built on quantum physics: probabilistic, discrete, measurement-dependent. The experiments in this notebook are the reason these two paradigms are different.

### Understanding Interference

Quantum algorithms work because of interference. Wrong answers cancel out. Right answers add up. This is directlyanalogous to wave interference. The dual wave-particle nature of quantum objects is what makes interference possible in quantum circuits.

### The Bottom Line

You do not need to understand every detail of every experiment. But you need to understand that quantum mechanics is not an arbitrary set of rules. It was forced upon us by experiments. Nature demanded it. The universe really works this way, whether we like it or not.

## Summary

Classical physics failed spectacularly when applied to atoms, light, and electrons:

- **Ultraviolet catastrophe**: Hot objects should emit infinite energy. They do not.
- **Photoelectric effect**: Light behaves like particles, not continuous waves.
- **Atomic spectra**: Atoms emit only specific frequencies, not continuous spectra.
- **Atom stability**: Classical atoms should collapse instantly. They do not.
- **Specific heat**: Drops mysteriously at low temperatures.
- **Compton scattering**: Photons bounce like particles with momentum.
- **Radioactive decay**: Fundamentally random, not determined.
- **Stern-Gerlach**: Angular momentum is quantized, not continuous.

The common theme: at the atomic scale, energy is discrete, randomness is fundamental, and the wave-particle distinction breaks down.
