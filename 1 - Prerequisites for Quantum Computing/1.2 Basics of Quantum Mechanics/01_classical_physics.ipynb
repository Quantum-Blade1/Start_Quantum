# Chapter 01: Classical Physics — The World Before Quantum

Before we can understand what quantum mechanics is, we need to understand what it replaced.

Classical physics is the set of rules that describe how the everyday world works. When you throw a ball, drive a car, watch the planets move, or listen to music, everything follows classical physics. It is the physics that Isaac Newton, James Clerk Maxwell, and many other brilliant minds built over hundreds of years.

Classical physics works beautifully for things we can see and touch. It predicted the motion of planets so well that astronomers could tell you exactly where Jupiter would be in 50 years. It explained how bridges hold weight, how engines generate power, and how light travels across the universe.

But here is the thing. Classical physics is wrong. Not a little wrong. It is fundamentally, completely wrong about how the universe works at the smallest scales. It just happens to be an incredibly good approximation for big things.

To understand why we needed quantum mechanics, you first need to understand what classical physics says, and why those ideas seemed so obviously correct that it took the greatest minds in history to realize they were not.

This notebook covers the key ideas from classical physics that quantum mechanics would later challenge and replace.

---

## 1. Newton's Mechanics: The Clockwork Universe

In the late 1600s, Isaac Newton published his laws of motion and his law of universal gravitation. These laws changed everything.

### Newton's Three Laws

**First Law (Inertia):** An object at rest stays at rest, and an object in motion stays in motion, unless something pushes or pulls on it.

Think about it this way. Imagine you are sitting in a car that suddenly brakes. Your body lurches forward. Why? Because your body was moving and it wanted to keep moving. Nothing told it to stop. The seat belt had to do that. That is inertia.

**Second Law (F = ma):** The force on an object equals its mass times its acceleration.

This is the most important equation in classical physics. Push something twice as hard, it accelerates twice as fast. Push something that is twice as heavy, it accelerates half as fast. Simple. Predictable. Beautiful.

Imagine pushing a shopping cart. An empty cart is easy to push — it accelerates quickly. A full cart is heavy — it accelerates slowly with the same push. That is F = ma.

**Third Law (Action and Reaction):** For every action, there is an equal and opposite reaction.

When you jump, you push the Earth down and the Earth pushes you up. When a rocket fires, it pushes exhaust gas backward and the gas pushes the rocket forward. Every force comes in pairs.

### The Clockwork Universe

Newton's laws led to a powerful idea: the universe is like a giant clock. If you know the position and velocity of every particle in the universe right now, you can predict everything that will ever happen. Every future event is determined by the present. There is no randomness, no uncertainty, no surprises.

The French mathematician Pierre-Simon Laplace put it this way: "An intellect which at a certain moment would know all forces and all positions of all items of which nature is composed — nothing would be uncertain and the future, just like the past, would be present before its eyes."

This idea is called **determinism**. In the classical world, if you know everything now, you know everything forever.

Imagine a perfect billiards player who knows the exact position, speed, and spin of every ball on the table. In the classical world, this player could predict exactly where every ball will end up after any shot. No guessing. No probability. Just pure, mechanical certainty.

Quantum mechanics destroys this idea completely. In the quantum world, even with perfect knowledge, the future is fundamentally uncertain. But we are getting ahead of ourselves.

## 2. Energy: The Currency of Physics

Energy is one of the most important concepts in all of physics. It is also one of the hardest to define simply. Energy is the ability to make things happen. When something moves, it has energy. When something is hot, it has energy. When something is high up, it has energy.

### Kinetic Energy

Kinetic energy is the energy of motion. A moving car has kinetic energy. A flying bird has kinetic energy. A spinning top has kinetic energy.

The formula is: **KE = (1/2) × mass × velocity²**

Notice that velocity is squared. This means speed matters much more than mass. A car going 100 km/h has four times the kinetic energy of the same car going 50 km/h. This is why car crashes at high speed are so much more destructive.

### Potential Energy

Potential energy is stored energy. A ball held above the ground has gravitational potential energy. A stretched rubber band has elastic potential energy. A charged battery has chemical potential energy.

Think of potential energy like money in a bank account. It is not doing anything right now, but it has the potential to do things when you spend it. When you drop the ball, the potential energy converts to kinetic energy.

### Conservation of Energy

Energy cannot be created or destroyed. It can only change forms. When you drop a ball, gravitational potential energy becomes kinetic energy. When the ball hits the ground and stops, the kinetic energy becomes heat and sound. The total amount of energy never changes.

This is the **law of conservation of energy**, and it is one of the most fundamental laws in all of physics. It holds true in classical physics. It holds true in quantum mechanics. It holds true in Einstein's relativity. It is never broken.

### Energy in Classical Physics is Continuous

This is the crucial point for our story. In classical physics, energy is **continuous**. This means energy can take any value. You can have 1 unit of energy, or 1.5 units, or 1.00001 units, or any number you like. There are no gaps. Energy flows smoothly, like water from a tap.

Imagine a ramp. In classical physics, energy is like walking up a ramp. You can stop at any height. There is no minimum step. You can go up by 1 centimeter, or 0.1 centimeters, or 0.0001 centimeters. Any amount.

Quantum mechanics says this is wrong. Energy comes in discrete chunks called **quanta**. It is not a ramp, it is a staircase. You can stand on step 1 or step 2, but never between them. This was the first crack in classical physics, and it changed everything. But that is for the next notebook.

## 3. Waves: Vibrations That Travel

Waves are one of the most important concepts in physics, and they are absolutely essential for understanding quantum mechanics later.

### What is a Wave?

A wave is a disturbance that travels through space and time. Drop a stone in a pond. The stone pushes water down, and that disturbance spreads outward as ripples. The water itself does not travel outward — it just bobs up and down. What travels is the pattern, the disturbance, the wave.

This is a key point. Waves carry energy and information without carrying matter. When you hear someone speak across a room, no air molecules travel from their mouth to your ear. Instead, the sound wave — a pattern of compressed and stretched air — travels across the room.

### Properties of Waves

Every wave has these properties:

**Wavelength (λ):** The distance between two consecutive peaks (or troughs) of the wave. Think of it as how long each wave cycle is.

**Frequency (f):** How many complete wave cycles pass a point per second. Measured in Hertz (Hz). If 10 waves pass by every second, the frequency is 10 Hz.

**Amplitude:** The height of the wave from the center line to the peak. Bigger amplitude means more energy. A loud sound has a bigger amplitude than a quiet sound. A bright light has a bigger amplitude than a dim light.

**Speed:** How fast the wave travels. For any wave: **speed = wavelength × frequency**.

### Interference: When Waves Meet

This is one of the most important properties of waves, and it comes back in a huge way in quantum mechanics.

When two waves meet, they do not bounce off each other like balls. They add together.

**Constructive interference:** When two wave peaks meet, they add up to make a bigger peak. Imagine two people on opposite sides of a rope both flicking it upward at the same time. The wave in the middle is twice as tall.

**Destructive interference:** When a peak meets a trough, they cancel each other out. The result can be zero — no wave at all. This is how noise-cancelling headphones work. They create a wave that is the exact opposite of the noise, and the two waves cancel out.

Interference is the signature property of waves. If something shows interference, it is behaving like a wave. If it does not, it is behaving like a particle. This distinction becomes incredibly important when we get to the double-slit experiment in notebook 04.

### Diffraction: Waves Bending Around Corners

When a wave passes through a small opening or around an obstacle, it bends and spreads out. This is called diffraction.

You can hear someone talking around a corner even though you cannot see them. That is because sound waves bend around corners (diffraction). Light waves also diffract, but because their wavelength is so tiny, the effect is only visible when the opening is extremely small.

Diffraction is another signature property of waves. And just like interference, it shows up in quantum mechanics in deeply surprising ways.

## 4. Electromagnetic Theory: Light as a Wave

In the 1860s, James Clerk Maxwell unified electricity and magnetism into a single theory. This was one of the greatest achievements in the history of physics.

### Electric and Magnetic Fields

A **field** is something that has a value at every point in space. Temperature is a field — every point in a room has a temperature. Wind is a field — every point in the atmosphere has a wind speed and direction.

An **electric field** exists around every charged object. If you put a positive charge near a negative charge, the electric field pulls them together. If you put two positive charges near each other, the electric field pushes them apart.

A **magnetic field** exists around every magnet and every electric current. It is what makes compass needles point north. It is what makes motors spin.

### Maxwell's Discovery: Light is an Electromagnetic Wave

Maxwell showed that electric and magnetic fields are deeply connected. A changing electric field creates a magnetic field. A changing magnetic field creates an electric field. These two fields can chase each other through space, creating a self-sustaining wave.

Maxwell calculated the speed of this electromagnetic wave. It came out to be exactly the speed of light: 300,000 kilometers per second.

This was not a coincidence. Light IS an electromagnetic wave.

This was a stunning discovery. For thousands of years, humans had wondered what light was. Now they knew: it is a wave of oscillating electric and magnetic fields flying through space.

### The Electromagnetic Spectrum

Light is just one type of electromagnetic wave. The full spectrum includes:

- **Radio waves** (longest wavelength, lowest frequency) — used for communication
- **Microwaves** — used for cooking and radar
- **Infrared** — heat radiation, used in night vision
- **Visible light** — the tiny sliver we can see (red, orange, yellow, green, blue, violet)
- **Ultraviolet** — causes sunburns
- **X-rays** — used in medical imaging
- **Gamma rays** (shortest wavelength, highest frequency) — emitted by nuclear reactions

All of these are the same thing: electromagnetic waves. They differ only in wavelength and frequency. A radio wave and a gamma ray are fundamentally the same phenomenon, just at vastly different scales.

### The Problem That Led to Quantum Mechanics

Maxwell's theory was incredibly successful. It explained light, radio waves, and electromagnetism beautifully. But it had a fatal flaw.

If light is a wave, and waves can have any energy (because energy is continuous in classical physics), then a hot object should radiate infinite energy at high frequencies. This is called the **ultraviolet catastrophe**, and it is one of the most important failures in the history of physics.

We will explore this in detail in notebook 02.

## 5. Thermodynamics: Heat and Entropy

Thermodynamics is the physics of heat, energy, and disorder. It is built on four laws.

### The Zeroth Law

If object A is in thermal equilibrium with object B, and object B is in thermal equilibrium with object C, then A and C are also in thermal equilibrium. This sounds obvious, but it is what allows us to define temperature consistently.

Think of it this way: if your coffee is the same temperature as the air, and the air is the same temperature as the thermometer, then the thermometer reads the temperature of your coffee. Temperature is transitive.

### The First Law

Energy is conserved. The total energy of an isolated system never changes. Heat can flow, work can be done, but the total energy stays the same.

This is just the conservation of energy applied to heat. When you boil water on a stove, the electrical energy from the stove becomes heat energy in the pot, which becomes kinetic energy of water molecules (making them move faster, which we experience as higher temperature).

### The Second Law

This is the big one. The entropy (disorder) of an isolated system never decreases. Things naturally go from ordered to disordered.

Drop a glass on the floor. It shatters into a hundred pieces. But a hundred glass pieces on the floor never spontaneously assemble into a glass. Why not? The second law. Going from ordered (glass) to disordered (broken pieces) is natural. Going the other way is not.

A room gets messily naturally but never cleans itself. An ice cube melts in warm water but warm water never spontaneously freezes into an ice cube. A perfume bottle opened in a room fills the room with scent, but the scent never spontaneously gathers back into the bottle.

The second law is one of the most profound laws in physics. It defines the direction of time. It tells us that the past is different from the future. And it connects deeply to information theory, which connects deeply to quantum computing.

### Statistical Mechanics: The Bridge

In the late 1800s, Ludwig Boltzmann showed that thermodynamics could be explained in terms of atoms and statistics. Heat is just atoms moving randomly. Temperature is the average kinetic energy of atoms. Entropy is the number of ways atoms can be arranged.

This was controversial because many physicists at the time did not even believe atoms existed. But Boltzmann was right.

Statistical mechanics is important for our story because it is where the quantum revolution began. When Boltzmann and others tried to apply statistical mechanics to light (electromagnetic radiation), they got nonsensical answers. The theory predicted that a hot object should emit infinite energy. Something was deeply wrong. The fix required inventing quantum mechanics.

## 6. Particles vs Waves: Two Types of Things

In classical physics, there are two fundamentally different types of things in the universe: particles and waves. They are completely separate categories.

### Particles

A particle is a localized object. It exists at one specific place at one specific time. It has a definite position and a definite velocity. You can track it.

Think of a tennis ball. At any given moment, it is at one specific location. It has one specific speed. It follows one specific path through the air. You can predict where it will be if you know its current position and velocity.

Properties of particles:
- They are localized (exist at a point)
- They follow definite paths (trajectories)
- When two particles collide, they bounce off each other
- They can be counted (one particle, two particles, three particles)
- They carry energy in concentrated lumps

### Waves

A wave is a spread-out disturbance. It does not exist at one point. It extends across space. You cannot point to one spot and say "the wave is right here." It is everywhere along its extent.

Think of an ocean wave. It is not at one point. It stretches across a long distance. It does not follow a single path. It spreads out, bends around obstacles, and passes through openings.

Properties of waves:
- They are delocalized (spread across space)
- They do not follow definite paths
- When two waves meet, they interfere (add or cancel)
- They cannot be counted in the traditional sense
- They carry energy spread out across space

### The Classical Rule: Pick One

In classical physics, everything is either a particle or a wave. Never both. A ball is a particle. A sound is a wave. An electron is a particle. Light is a wave.

The question "is light a particle or a wave?" had been debated for centuries. Newton said particle. Huygens said wave. By the 1800s, experiments on interference and diffraction had settled the debate: light is definitely a wave.

Or so everyone thought.

Quantum mechanics would prove that this neat separation is wrong. Things can be both particle and wave at the same time. But in the classical world, the separation seemed absolute and obvious.

## 7. The Classical Atom

By the late 1800s and early 1900s, scientists had discovered that matter is made of atoms, and atoms are made of smaller particles.

### The Discovery of the Electron

In 1897, J.J. Thomson discovered the electron by studying cathode rays (beams of particles in vacuum tubes). He showed that these particles were much lighter than atoms and carried a negative charge. This meant atoms were not the smallest things. They had parts.

### Thomson's Plum Pudding Model

Thomson proposed that an atom was like a pudding (or a sphere) of positive charge with electrons embedded in it like plums in a pudding. The positive charge was spread out uniformly, and the negative electrons sat inside it. This model was simple and made sense.

Imagine a ball of cookie dough (positive charge) with chocolate chips (electrons) scattered throughout it. That was Thomson's picture of the atom.

### Rutherford's Discovery: The Nucleus

In 1909, Ernest Rutherford's team fired alpha particles (small, positively charged particles) at a thin gold foil. Most alpha particles passed straight through. But some bounced back at large angles. A few even bounced straight back.

Rutherford said this was as surprising as firing a cannon at tissue paper and having the cannonball bounce back at you.

The conclusion: the atom is mostly empty space. Almost all the mass and positive charge is concentrated in a tiny, dense nucleus at the center. The electrons orbit around this nucleus at a great distance, like planets orbiting the sun.

### The Problem with Rutherford's Atom

Rutherford's model had a fatal flaw. According to Maxwell's electromagnetic theory, a charged particle that accelerates (and orbiting is a form of acceleration — the direction is constantly changing) must radiate electromagnetic waves. This means the orbiting electron should continuously emit light, lose energy, spiral inward, and crash into the nucleus.

According to classical physics, every atom in the universe should collapse in about one hundred-trillionth of a second.

Obviously, atoms do not collapse. We exist. Something is wrong with classical physics.

This was one of the most serious problems in physics at the turn of the 20th century. The solution required completely new ideas about energy, light, and the behavior of electrons. It required quantum mechanics.

## 8. Fields: The Fabric of Reality

Fields are one of the most important concepts in modern physics, and they play a central role in quantum mechanics.

### What is a Field?

A field is a quantity that has a value at every single point in space. You cannot escape a field. It is everywhere.

**Gravitational field:** Every point around the Earth has a gravitational field value. This field tells any mass placed at that point how strong the gravitational pull is and which direction it points. You do not need to touch the Earth to feel its gravity. The field carries the influence.

**Temperature field:** Every point in a room has a temperature. Near the heater, the temperature is high. Near the window, it is low. The temperature varies smoothly from point to point. That is a field.

**Electric field:** Every point near a charged object has an electric field. This field tells other charges how they would be pushed or pulled if placed at that point.

### Why Fields Matter

Before fields, physicists thought that forces required contact. How does the sun pull on the Earth when they are 150 million kilometers apart? Newton had no answer. He called it "action at a distance" and admitted it was philosophically troubling.

Fields solve this problem. The sun does not directly pull on the Earth. The sun creates a gravitational field everywhere in space. The Earth, sitting in that field, feels the pull. The influence travels through the field, not through empty space.

### Fields in Quantum Mechanics

In quantum mechanics, fields become even more important. The wave function, which describes a quantum system, is a type of field. It has a value at every point in space. Quantum field theory (the deepest version of quantum mechanics) says that everything in the universe — every particle, every force — is actually a vibration in a field.

An electron is not a little ball. It is a vibration in the electron field. A photon is not a little bullet of light. It is a vibration in the electromagnetic field. This is the modern view of physics.

But for now, just remember: a field is something that exists everywhere in space and has a value at every point. You will see this idea over and over in the notebooks that follow.

## 9. Conservation Laws: The Unchanging Rules

Classical physics discovered several conservation laws. These are quantities that never change, no matter what happens.

### Conservation of Energy

We already discussed this. Energy can change forms but the total amount never changes.

### Conservation of Momentum

Momentum is mass times velocity. In a closed system, the total momentum never changes.

When a billiard ball hits another ball, the first ball slows down and the second speeds up. The total momentum of both balls before the collision equals the total momentum after. Always.

When a gun fires, the bullet goes forward (gaining momentum) and the gun recoils backward (gaining momentum in the opposite direction). The total momentum is zero before and after.

### Conservation of Angular Momentum

Angular momentum is the rotational version of momentum. A spinning top keeps spinning unless friction slows it down. An ice skater who pulls in their arms spins faster — the angular momentum stays the same, but because the mass is closer to the center, the spin rate must increase.

### Conservation of Charge

Electric charge is conserved. You cannot create or destroy charge. You can move it around, but the total amount of positive charge minus negative charge in the universe never changes.

### Why Conservation Laws Matter for Quantum Mechanics

All of these conservation laws survive in quantum mechanics. Energy, momentum, angular momentum, and charge are all conserved in the quantum world too. These laws are so deep that they transcend the specific theory. Whether you use classical physics or quantum physics, conservation laws hold.

In fact, the physicist Emmy Noether proved one of the most beautiful theorems in all of physics: every conservation law corresponds to a symmetry of nature. Energy conservation comes from time symmetry (the laws of physics do not change from one moment to the next). Momentum conservation comes from space symmetry (the laws of physics are the same everywhere). Angular momentum conservation comes from rotational symmetry.

This connection between symmetry and conservation is one of the deepest ideas in all of physics, and it carries directly into quantum mechanics and quantum computing.

## 10. The State of Physics by 1900

By the year 1900, classical physics seemed nearly complete. The famous physicist Lord Kelvin reportedly said that physics was essentially done, with only "two small clouds" on the horizon.

Those two clouds were:

1. **The failure to detect the luminiferous ether** — the medium through which light waves were supposed to travel. The Michelson-Morley experiment had failed to find it. This cloud led to Einstein's special relativity.

2. **The ultraviolet catastrophe** — the prediction that hot objects should emit infinite energy at high frequencies. This cloud led to quantum mechanics.

Both "small clouds" turned out to be enormous storms that completely revolutionized physics.

### What Classical Physics Got Right

- Mechanics (Newton's laws)
- Electromagnetism (Maxwell's equations)
- Thermodynamics and statistical mechanics
- Optics (light as a wave)
- Gravity (Newton's law of gravitation)
- Conservation laws

### The Assumptions That Turned Out To Be Wrong

1. **Energy is continuous.** You can have any amount of energy. There is no minimum step.
   → Quantum mechanics says: energy comes in discrete chunks (quanta).

2. **Particles and waves are separate things.** Something is either a particle or a wave, never both.
   → Quantum mechanics says: everything is both, depending on how you look at it.

3. **Measurement does not affect the system.** You can observe something without changing it.
   → Quantum mechanics says: measurement fundamentally changes the system.

4. **The universe is deterministic.** If you know the present perfectly, you can predict the future perfectly.
   → Quantum mechanics says: the future is fundamentally probabilistic. Even with perfect knowledge, you cannot predict outcomes with certainty.

5. **Objects have definite properties at all times.** A ball has a definite position and velocity whether or not anyone is looking.
   → Quantum mechanics says: quantum objects do not have definite properties until they are measured.

6. **Information can be freely copied.** You can look at a document and make a perfect copy.
   → Quantum mechanics says: quantum information cannot be copied (the no-cloning theorem).

### The Stage is Set

Everything we have discussed in this notebook — Newton's mechanics, energy, waves, electromagnetic theory, thermodynamics, particles, fields, conservation laws — is the foundation that quantum mechanics was built on top of. Quantum mechanics did not throw all of this away. It modified it, extended it, and in some places completely replaced it.

In the next notebook, we will see exactly where classical physics broke. We will examine the specific experiments that produced results that classical physics simply could not explain. These are the experiments that forced physicists to invent an entirely new theory of reality: quantum mechanics.

## Summary

Classical physics describes the everyday world with remarkable accuracy:

- **Newton's mechanics** gives us deterministic, clockwork physics where everything is predictable
- **Energy** can take any value (continuous) and is always conserved
- **Waves** spread out, interfere, and diffract
- **Particles** are localized and follow definite paths
- **Maxwell's electromagnetism** shows that light is a wave
- **Thermodynamics** governs heat, entropy, and the arrow of time
- **Fields** carry forces through space
- **Conservation laws** (energy, momentum, charge) are unbreakable

But by 1900, cracks were appearing. Classical physics could not explain:
- Why atoms are stable
- Why hot objects do not emit infinite energy
- Why light behaves like particles in some experiments
- Why electrons in atoms only have certain energies

The next notebook explores these failures in detail — the experiments that broke classical physics and forced humanity to discover quantum mechanics.
