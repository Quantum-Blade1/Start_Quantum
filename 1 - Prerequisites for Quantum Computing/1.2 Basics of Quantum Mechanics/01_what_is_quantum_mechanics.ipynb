# 01 - What is Quantum Mechanics?

Quantum mechanics is a set of rules that describe how very small things behave. Atoms, electrons, photons -- they do not follow the same rules as a ball or a car. They follow quantum rules.

For quantum computing, you do not need to learn all of quantum mechanics. You need to learn the parts that explain how qubits work. That is what this notebook covers.

There are four main postulates (rules) of quantum mechanics that matter for quantum computing:

1. **State Space** -- How we describe the state of a quantum system
2. **Evolution** -- How quantum states change over time
3. **Measurement** -- What happens when we look at a quantum system
4. **Composition** -- How we combine multiple quantum systems

## Classical vs Quantum: The Key Differences

| Property | Classical | Quantum |
|----------|-----------|--------|
| State | Definite (0 or 1) | Superposition (both until measured) |
| Measurement | Does not change the system | Collapses the state |
| Copying | Can copy freely | Cannot copy (no-cloning theorem) |
| Correlation | Independent bits | Entangled qubits |
| Information | Bits (0 or 1) | Qubits (complex vector in Hilbert space) |

```python
import numpy as np

# Classical bit: either 0 or 1
classical_bit = 1
print(f"Classical bit: {classical_bit}")
print(f"It is definitely {classical_bit}. No ambiguity.\n")

# Quantum bit: a vector with complex amplitudes
# This qubit has a 70% chance of being 0 and 30% chance of being 1
qubit = np.array([np.sqrt(0.7), np.sqrt(0.3)], dtype=complex)
print(f"Qubit state: {qubit.round(4)}")
print(f"P(0) = {abs(qubit[0])**2:.2f}")
print(f"P(1) = {abs(qubit[1])**2:.2f}")
print("Before measurement, it is BOTH. Not unknown, not hidden -- genuinely both.")
```

## Wave-Particle Duality

Quantum objects behave like waves and like particles at the same time. This sounds strange but it is an experimental fact.

- As a **particle**: when you measure an electron, you find it at a specific location
- As a **wave**: before measurement, it exists as a probability wave spread across space

For quantum computing, the wave part is important because waves can **interfere**. Quantum algorithms work by making wrong answers cancel out (destructive interference) and right answers add up (constructive interference).

```python
# Interference example using Hadamard gate
# H|0> = |+>   (equal superposition)
# H|+> = |0>   (interference brings it back)

H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
ket_0 = np.array([1, 0], dtype=complex)

# Step 1: Create superposition
after_H1 = H @ ket_0
print("After first H (superposition):")
print(f"  State: {after_H1.round(4)}")
print(f"  P(0) = {abs(after_H1[0])**2:.2f}, P(1) = {abs(after_H1[1])**2:.2f}")

# Step 2: Apply H again -- interference happens
after_H2 = H @ after_H1
print("\nAfter second H (interference):")
print(f"  State: {after_H2.round(4)}")
print(f"  P(0) = {abs(after_H2[0])**2:.2f}, P(1) = {abs(after_H2[1])**2:.2f}")

print("\nThe two paths to |1> cancelled out (destructive interference).")
print("The two paths to |0> added up (constructive interference).")
print("This is how quantum algorithms amplify correct answers.")
```

## The Uncertainty Principle

You cannot know everything about a quantum system at the same time. This is not a limitation of our instruments. It is a fundamental property of nature.

For quantum computing, this shows up as: you cannot measure a qubit in two different bases simultaneously. If you know the Z-basis value perfectly, you know nothing about the X-basis value, and vice versa.

```python
# |0> in Z-basis: definitely 0
ket_0 = np.array([1, 0], dtype=complex)

# But in X-basis (|+>, |->): completely uncertain
ket_plus = np.array([1, 1], dtype=complex) / np.sqrt(2)
ket_minus = np.array([1, -1], dtype=complex) / np.sqrt(2)

print("State |0> measured in Z-basis:")
print(f"  P(|0>) = {abs(np.dot(ket_0.conj(), ket_0))**2:.2f} -- certain")
print(f"  P(|1>) = {abs(np.dot(np.array([0,1]), ket_0))**2:.2f}")

print("\nSame state |0> measured in X-basis:")
print(f"  P(|+>) = {abs(np.dot(ket_plus.conj(), ket_0))**2:.2f} -- uncertain")
print(f"  P(|->) = {abs(np.dot(ket_minus.conj(), ket_0))**2:.2f} -- uncertain")

print("\nPerfect knowledge in one basis means zero knowledge in the other.")
print("This is the uncertainty principle applied to qubits.")
```

## Key Takeaway

Quantum mechanics for quantum computing boils down to:
- States are vectors (superposition is real, not just ignorance)
- Measurement changes the state (wave function collapse)
- Interference makes quantum algorithms work
- You cannot copy quantum states or measure everything at once
