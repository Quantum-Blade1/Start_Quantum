# Chapter 03: Inner Products and Norms — The Geometry

Welcome to **Section 1.1: Essential Linear Algebra**.

We have complex numbers (the values). We have vector spaces (the stage). But something is missing.

A vector space by itself is just a collection of arrows. It doesn't tell you:
- How long an arrow is (Norm).
- The angle between two arrows.
- Whether two arrows are perpendicular (Orthogonal).

To answer these questions, we need to equip our vector space with a new tool: the **Inner Product**.

This turns our boring vector space into a **Hilbert Space**.

In quantum mechanics, the inner product is everything. It tells us probabilities. It tells us if states are distinguishable. It tells us if an operation is valid.

---

## 1. The Dot Product (A Familiar Friend)

You probably remember the dot product from school. For real vectors $\vec{a}$ and $\vec{b}$:

$$ \vec{a} \cdot \vec{b} = a_1 b_1 + a_2 b_2 + ... + a_n b_n $$

It also has a geometric meaning:
$$ \vec{a} \cdot \vec{b} = |\vec{a}| |\vec{b}| \cos(\theta) $$

This connects algebra (multiplying numbers) to geometry (angles).

But in quantum mechanics, we use complex numbers. And for complex numbers, the standard dot product breaks.

---

## 2. The Inner Product (Complex Version)

Why does it break? Because the length of a vector $|v|^2$ should be $\vec{v} \cdot \vec{v}$.
If $v = i$, then $v \cdot v = i \times i = -1$.
A length squared of -1? That's impossible. Lengths must be positive.

**The Fix:** We use the **Conjugate**.

The inner product of two complex vectors $|u\rangle$ and $|v\rangle$ is defined as:

$$ \langle u | v \rangle = u_1^* v_1 + u_2^* v_2 + ... + u_n^* v_n $$

Notice the star ($^*$) on the first vector ($u$). We take the conjugate of the first vector, and leave the second one alone.

This small change fixes everything.
Now, the inner product of a vector with itself is:
$$ \langle v | v \rangle = v_1^* v_1 + v_2^* v_2 + ... $$
$$ = |v_1|^2 + |v_2|^2 + ... $$

This is always a real, non-negative number. It represents the **Norm squared**.

```python
import numpy as np

# Defining complex vectors
u = np.array([1 + 1j, 2])
v = np.array([3, 1j])

# Calculating Inner Product <u|v>
# We must conjugate u first!
inner_prod = np.dot(u.conj(), v)
# Manual check: (1-i)*3 + (2)*i = 3 - 3i + 2i = 3 - i

print(f"Inner product <u|v>: {inner_prod}")

# Note: Numpy has a dedicated function np.vdot() which handles this automatically.
# BEWARE: np.vdot(a, b) conjugates the FIRST argument (a). This matches physics notation.
inner_prod_vdot = np.vdot(u, v)
print(f"Using np.vdot: {inner_prod_vdot}")
```

## 3. Norm and Normalization

The **Norm** (length) of a vector $|\psi\rangle$ is:

$$ ||\psi|| = \sqrt{\langle \psi | \psi \rangle} $$

In quantum mechanics, vectors represent probabilities. Since total probability must be 1 (100%), all valid explicit quantum states must have a norm of 1.

This is called being **Normalized**.

If you have a vector that is not normalized (e.g., length 5), you can normalize it by dividing by its length.

$$ |\psi_{normalized}\rangle = \frac{|\psi\rangle}{||\psi||} $$

```python
vec = np.array([3, 4])

# Calculate norm
norm = np.linalg.norm(vec)
print(f"Original vector: {vec}")
print(f"Norm: {norm}") # Should be 5

# Normalize
normalized_vec = vec / norm
print(f"Normalized vector: {normalized_vec}")

# Verify new norm is 1
print(f"New norm: {np.linalg.norm(normalized_vec)}")
```

## 4. Orthogonality: The Zero Overlap

Two vectors are **Orthogonal** (perpendicular) if their inner product is **zero**.

$$ \langle u | v \rangle = 0 $$

This is a massive concept.
Geometrically, means they are at 90 degrees.
Informationally, it means they are **perfectly distinguishable**.

If $|0\rangle$ and $|1\rangle$ are orthogonal (which they are), it means if you have a state that is definitely $|0\rangle$, it has absolutely zero "overlap" with $|1\rangle$. You will never mistake one for the other.

It also means they are independent events.

### Orthonormal Basis
A basis is **Orthonormal** if:
1. All vectors are normalized (length 1).
2. All vectors are mutually orthogonal (overlap 0).

The standard basis $\{|0\rangle, |1\rangle\}$ is orthonormal.
$\langle 0 | 0 \rangle = 1$
$\langle 1 | 1 \rangle = 1$
$\langle 0 | 1 \rangle = 0$

This makes calculations incredibly easy.

```python
# Checking orthogonality
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

overlap = np.vdot(ket_0, ket_1)
print(f"<0|1>: {overlap}")

# What about H basis?
plus = (1/np.sqrt(2)) * (ket_0 + ket_1)
minus = (1/np.sqrt(2)) * (ket_0 - ket_1)

overlap_pm = np.vdot(plus, minus)
print(f"<+|- >: {overlap_pm}")
# (1/2) * (1 - 1) = 0. They are orthogonal too!
```

## 5. The Cauchy-Schwarz Inequality

This is the most famous inequality in mathematics.

$$ |\langle u | v \rangle|^2 \leq \langle u | u \rangle \langle v | v \rangle $$

Or simpler: $|\langle u | v \rangle| \leq ||u|| \cdot ||v||$.

It says: The inner product of two vectors can never be larger than the product of their lengths.

When does equality hold? Only when $|u\rangle$ and $|v\rangle$ point in the exact same direction (they are parallel). $u = c v$.

This gives us a geometric interpretation of the inner product:
It measures **"how parallel"** two states are. It measures their overlap.

- Overlap 0: Orthogonal (Perpendicular).
- Overlap 1 (assuming normalized): Identical (Parallel).
- Overlap in between: Some probability of transitioning from one to the other.

---

## 6. Projecting with Inner Products

One final intuition.

Calculating $\langle 0 | \psi \rangle$ is asking: "How much of the state $|\psi\rangle$ lies along the direction of $|0\rangle$?"

If $|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$, then:
$$ \langle 0 | \psi \rangle = \langle 0 | (\alpha|0\rangle + \beta|1\rangle) $$
$$ = \alpha\langle 0 | 0 \rangle + \beta\langle 0 | 1 \rangle $$
$$ = \alpha(1) + \beta(0) $$
$$ = \alpha $$

See? The inner product **picks out the coefficient** $\alpha$. It projects the state onto the basis vector.

This is exactly what happens during measurement. We project onto the computational basis, and the probability is the square of this projection ($|\alpha|^2$).
