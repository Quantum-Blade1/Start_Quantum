# Inner Products and Norms: Measuring Quantum Overlap

## The Inner Product: Quantum Overlap

The **inner product** ⟨v|w⟩ measures how much two quantum states "look alike":

$$\langle v | w \rangle = \sum_i v_i^* w_i = v_1^* w_1 + v_2^* w_2 + \cdots$$

**Critical**: Complex conjugate on FIRST vector. This asymmetry ensures ⟨v|v⟩ is real and positive.

**Properties**:
- Conjugate symmetry: ⟨v|w⟩ = ⟨w|v⟩*
- Linearity in second: ⟨v|(αu + βw)⟩ = α⟨v|u⟩ + β⟨v|w⟩
- Anti-linearity in first: ⟨(αu + βv)|w⟩ = α*⟨u|w⟩ + β*⟨v|w⟩
- Positive-definite: ⟨v|v⟩ ≥ 0, equality iff v = 0

## Probability and Measurement

**The measurement postulate**: Probability of measuring eigenstate |e⟩ when in state |ψ⟩:
$$P(e) = |\langle e | \psi \rangle|^2 = \langle e | \psi \rangle^* \langle e | \psi \rangle$$

This is EVERYTHING. The inner product computes measurement probabilities—it's the bridge from abstract quantum state to physical outcomes.

Example: State |ψ⟩ = (|0⟩ + |1⟩)/√2 measured in computational basis:
$$P(0) = |\langle 0|\psi\rangle|^2 = |1/\sqrt{2}|^2 = 1/2$$
$$P(1) = |\langle 1|\psi\rangle|^2 = |1/\sqrt{2}|^2 = 1/2$$

## Norms and Normalization

The **norm** measures vector length:
$$\|\psi\| = \sqrt{\langle \psi | \psi \rangle}$$

For quantum states: **Normalized** means ‖ψ‖ = 1

Why? Because probabilities must sum to 1:
$$\sum_i P(i) = \sum_i |\langle i|\psi\rangle|^2 = \langle \psi | \psi \rangle = \|\psi\|^2$$

This requires ‖ψ‖² = 1.

## Orthogonality: Distinguishable States

States u and v are **orthogonal** if ⟨u|v⟩ = 0 (zero overlap, completely different).

**Orthonormal basis**: Each basis vector has norm 1 and orthogonal to others:
$$\langle e_i | e_j \rangle = \delta_{ij} = \begin{cases} 1 & i = j \\ 0 & i \neq j \end{cases}$$

Computational basis IS orthonormal:
$$\langle 0|0 \rangle = 1, \; \langle 1|1 \rangle = 1, \; \langle 0|1 \rangle = 0$$

This orthogonality is why measurements in computational basis give definite outcomes (0 or 1). Non-orthogonal states are indistinguishable by measurement.

## Cauchy-Schwarz Inequality

$$|\langle v | w \rangle| \leq \|v\| \cdot \|w\|$$

**Quantum meaning**: The overlap between states is bounded. This inequality guarantees measurement probabilities are ≤ 1.

```python
# Functional code: Inner products and probability

# Orthonormal computational basis
ket_0 = np.array([1, 0], dtype=complex)
ket_1 = np.array([0, 1], dtype=complex)

# Arbitrary normalized state
psi = np.array([0.6, 0.8], dtype=complex)  # normalized

# Inner products = measurement amplitudes
amplitude_0 = np.vdot(ket_0, psi)  # ⟨0|ψ⟩
amplitude_1 = np.vdot(ket_1, psi)  # ⟨1|ψ⟩

# Measurement probabilities
prob_0 = abs(amplitude_0)**2
prob_1 = abs(amplitude_1)**2

# Orthogonality check
overlap = np.vdot(ket_0, ket_1)  # should be 0

# Gram-Schmidt orthonormalization
v1 = np.array([1, 1], dtype=complex)
e1 = v1 / np.linalg.norm(v1)
v2 = np.array([1, -1], dtype=complex)
v2 = v2 - np.dot(np.conj(e1), v2) * e1
e2 = v2 / np.linalg.norm(v2)
```

## Projection Operators and Measurement

The **outer product** |v⟩⟨w| creates a matrix projecting onto |v⟩ in direction |w⟩:

$$P = |v\rangle \langle v|$$

When measuring |ψ⟩ to determine if it's in state |v⟩:
$$P(\text{get } v) = \langle \psi | (|v\rangle \langle v|) |\psi\rangle = |\langle v|\psi\rangle|^2$$

Projection operators are ESSENTIAL for understanding quantum measurement.

## The Hermitian Structure

For measurement operators to make physical sense, they must be **Hermitian**: A† = A

This forces:
- Real eigenvalues (measurement outcomes must be real!)
- Orthogonal eigenvectors (different outcomes are orthogonal states)

Inner products enable this: ⟨ψ|A|ψ⟩ is always real when A is Hermitian because:
$$\langle\psi|A|\psi\rangle^* = \langle\psi|A†|\psi\rangle = \langle\psi|A|\psi\rangle$$
