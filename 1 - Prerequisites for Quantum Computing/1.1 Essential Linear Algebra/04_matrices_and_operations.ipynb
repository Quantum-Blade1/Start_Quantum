# 04 - Matrices and Matrix Operations

Every quantum gate is a matrix. When you apply an X gate to a qubit, you are multiplying a 2x2 matrix by a 2x1 vector. When you apply a CNOT gate to two qubits, you are multiplying a 4x4 matrix by a 4x1 vector. That is all quantum computation is at the math level -- matrix-vector multiplication.

```python
import numpy as np
```

## Creating Matrices

A matrix is a 2D array of numbers. In quantum computing, we mostly deal with square matrices (same number of rows and columns).

```python
# A 2x2 matrix
A = np.array([[1, 2],
              [3, 4]], dtype=complex)
print("Matrix A:")
print(A)

# The Pauli X gate (quantum NOT gate)
X = np.array([[0, 1],
              [1, 0]], dtype=complex)
print("\nPauli X gate:")
print(X)
```

## Matrix Addition and Scalar Multiplication

These work element by element, just like with vectors.

```python
A = np.array([[1, 2], [3, 4]], dtype=complex)
B = np.array([[5, 6], [7, 8]], dtype=complex)

# Addition
print("A + B =")
print(A + B)

# Scalar multiplication
print("\n3 * A =")
print(3 * A)
```

## Matrix Multiplication

This is the big one. Matrix multiplication is NOT element-by-element. The order matters: A * B is NOT the same as B * A.

In quantum computing, applying gate A first and then gate B means you compute B * A * |state>. Read right to left.

```python
A = np.array([[1, 2], [3, 4]], dtype=complex)
B = np.array([[5, 6], [7, 8]], dtype=complex)

# Matrix multiplication (use @ operator)
print("A @ B =")
print(A @ B)

print("\nB @ A =")
print(B @ A)

print("\nA @ B is NOT equal to B @ A:", not np.array_equal(A @ B, B @ A))
```

## Matrix Acting on a Vector (Applying a Gate)

This is the core operation in quantum computing. A gate (matrix) acts on a qubit state (vector) to produce a new state.

```python
# Pauli X gate (bit flip)
X = np.array([[0, 1],
              [1, 0]], dtype=complex)

ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

# Apply X to |0> -- should give |1>
result = X @ ket_0
print("X|0> =")
print(result)  # [0, 1] = |1>

# Apply X to |1> -- should give |0>
result = X @ ket_1
print("\nX|1> =")
print(result)  # [1, 0] = |0>

# X gate flips |0> to |1> and |1> to |0>. It is the quantum NOT gate.
```

## Identity Matrix

The identity matrix does nothing. It is the "do nothing" gate. Multiplying any vector by the identity gives back the same vector.

```python
I = np.eye(2, dtype=complex)  # 2x2 identity
print("Identity matrix:")
print(I)

ket_0 = np.array([[1], [0]], dtype=complex)
print("\nI|0> =")
print(I @ ket_0)  # Same as |0>
```

## Transpose, Conjugate, and Conjugate Transpose (Dagger)

Three operations you need to know:

1. **Transpose** (A^T): Swap rows and columns
2. **Complex conjugate** (A*): Flip sign of imaginary parts
3. **Conjugate transpose / Dagger** (A^dagger): Do both. This is the most important one in quantum computing.

```python
A = np.array([[1+1j, 2-1j],
              [3+0j, 4+2j]], dtype=complex)

print("Original A:")
print(A)

# Transpose
print("\nTranspose A^T:")
print(A.T)

# Complex conjugate
print("\nConjugate A*:")
print(np.conj(A))

# Conjugate transpose (dagger) -- THE important one
print("\nConjugate transpose A^dagger:")
print(A.conj().T)
```

## Applying Multiple Gates (Matrix Multiplication Chain)

If you apply gate A first, then gate B, then gate C, the combined operation is C @ B @ A.

Read right to left: A acts first, then B, then C.

```python
# Hadamard gate
H = np.array([[1, 1],
              [1, -1]], dtype=complex) / np.sqrt(2)

# Pauli X gate
X = np.array([[0, 1],
              [1, 0]], dtype=complex)

ket_0 = np.array([[1], [0]], dtype=complex)

# Apply H first, then X
# Step 1: H|0> = |+>
after_H = H @ ket_0
print("After H:")
print(after_H)

# Step 2: X|+>
after_XH = X @ after_H
print("\nAfter X(H):")
print(after_XH)

# Or do it in one step: (X @ H) @ |0>
combined = X @ H
result = combined @ ket_0
print("\nCombined (X @ H)|0>:")
print(result)  # Same answer
```

## Key Takeaway

- Quantum gates = matrices
- Applying a gate = matrix times vector
- Multiple gates = multiply matrices (right to left order)
- Conjugate transpose (dagger) is used everywhere
- Matrix multiplication order matters (not commutative)
