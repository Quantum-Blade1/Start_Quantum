# 01 - Complex Numbers

Qubit states use complex numbers as amplitudes. A qubit state |psi> = a|0> + b|1> where a and b are complex numbers. If you do not understand complex numbers, you cannot understand qubits. Simple as that.

This notebook covers what you actually need, nothing extra.

## What is a Complex Number?

A complex number has two parts: a real part and an imaginary part.

Written as: `z = a + bi`

Where:
- `a` is the real part
- `b` is the imaginary part
- `i` is the square root of -1 (Python uses `j` instead of `i`)

That is it. Nothing mysterious about it. It is just a number with two components instead of one.

```python
import numpy as np

# Creating complex numbers in Python
z1 = 3 + 4j
z2 = 1 - 2j

print(f"z1 = {z1}")
print(f"z2 = {z2}")
print(f"Real part of z1: {z1.real}")
print(f"Imaginary part of z1: {z1.imag}")
```

## Basic Operations

You can add, subtract, multiply, and divide complex numbers just like regular numbers. Python handles it for you.

```python
z1 = 3 + 4j
z2 = 1 - 2j

# Addition
print(f"z1 + z2 = {z1 + z2}")

# Subtraction
print(f"z1 - z2 = {z1 - z2}")

# Multiplication
# (3+4j)(1-2j) = 3 - 6j + 4j - 8j^2 = 3 - 2j + 8 = 11 - 2j
print(f"z1 * z2 = {z1 * z2}")

# Division
print(f"z1 / z2 = {z1 / z2}")
```

## Complex Conjugate

The conjugate of `a + bi` is `a - bi`. You just flip the sign of the imaginary part.

**Why it matters in quantum computing:** When you compute the probability of measuring a qubit state, you multiply the amplitude by its conjugate. The probability of measuring state |0> is `a * conjugate(a) = |a|^2`.

```python
z = 3 + 4j

# Conjugate
z_conj = np.conj(z)
print(f"z = {z}")
print(f"conjugate(z) = {z_conj}")

# Multiply a number by its conjugate -- always gives a real number
print(f"z * conjugate(z) = {z * z_conj}")
# Notice: (3+4j)(3-4j) = 9 + 16 = 25 (real number, no imaginary part)
```

## Modulus (Absolute Value)

The modulus of `a + bi` is `sqrt(a^2 + b^2)`. It is the "length" of the complex number.

**Why it matters:** In quantum computing, `|amplitude|^2` gives the probability of a measurement outcome. The modulus squared of all amplitudes in a qubit state must add up to 1. This is called the normalization condition.

```python
z = 3 + 4j

# Modulus
mod_z = abs(z)  # or np.abs(z)
print(f"|z| = {mod_z}")  # sqrt(9 + 16) = sqrt(25) = 5

# Modulus squared -- this is what gives probability in quantum computing
print(f"|z|^2 = {mod_z**2}")

# Example: a qubit state |psi> = (1/sqrt(2))|0> + (1/sqrt(2))|1>
a = 1/np.sqrt(2)  # amplitude for |0>
b = 1/np.sqrt(2)  # amplitude for |1>

prob_0 = abs(a)**2
prob_1 = abs(b)**2

print(f"\nQubit state |+>:")
print(f"Probability of measuring |0>: {prob_0}")
print(f"Probability of measuring |1>: {prob_1}")
print(f"Total probability: {prob_0 + prob_1}")  # Must be 1.0
```

## Euler's Formula and Polar Form

Any complex number can be written as:

`z = r * e^(i*theta)`

Where:
- `r` is the modulus (length)
- `theta` is the angle (phase)

Euler's formula: `e^(i*theta) = cos(theta) + i*sin(theta)`

**Why it matters:** In quantum computing, the "phase" of a qubit is this angle theta. Phase is invisible when you measure, but it affects how qubits interfere with each other. This is the entire basis of quantum algorithms.

```python
# Polar form: z = r * e^(i*theta)
z = 3 + 4j

r = abs(z)                    # modulus
theta = np.angle(z)           # phase angle in radians

print(f"z = {z}")
print(f"r (modulus) = {r}")
print(f"theta (phase) = {theta:.4f} radians = {np.degrees(theta):.2f} degrees")

# Reconstruct from polar form
z_reconstructed = r * np.exp(1j * theta)
print(f"\nReconstructed: {z_reconstructed.real:.4f} + {z_reconstructed.imag:.4f}j")

# Euler's formula: e^(i*pi) = -1 (the most beautiful equation in math)
euler = np.exp(1j * np.pi)
print(f"\ne^(i*pi) = {euler.real:.4f} + {euler.imag:.4f}j")  # Should be -1
```

## Complex Numbers in Qubit States

Now let us see how all of this actually shows up in quantum computing.

A single qubit state is: `|psi> = alpha|0> + beta|1>`

Where alpha and beta are complex numbers, and `|alpha|^2 + |beta|^2 = 1`.

Let us build some qubit states and verify the math.

```python
# State |0> -- 100% chance of measuring 0
state_0 = np.array([1, 0], dtype=complex)
print("State |0>:", state_0)
print(f"  P(0) = {abs(state_0[0])**2}, P(1) = {abs(state_0[1])**2}")

# State |1> -- 100% chance of measuring 1
state_1 = np.array([0, 1], dtype=complex)
print("\nState |1>:", state_1)
print(f"  P(0) = {abs(state_1[0])**2}, P(1) = {abs(state_1[1])**2}")

# State |+> -- 50/50 chance (created by Hadamard gate)
state_plus = np.array([1/np.sqrt(2), 1/np.sqrt(2)], dtype=complex)
print("\nState |+>:", state_plus)
print(f"  P(0) = {abs(state_plus[0])**2:.4f}, P(1) = {abs(state_plus[1])**2:.4f}")

# A state with complex amplitudes
# |psi> = (1/sqrt(2))|0> + (i/sqrt(2))|1>
state_complex = np.array([1/np.sqrt(2), 1j/np.sqrt(2)], dtype=complex)
print("\nState with complex amplitude:", state_complex)
print(f"  P(0) = {abs(state_complex[0])**2:.4f}, P(1) = {abs(state_complex[1])**2:.4f}")
print(f"  Total = {abs(state_complex[0])**2 + abs(state_complex[1])**2:.4f}")
# Still 50/50 even though the amplitude is imaginary
# The PHASE is different, but the PROBABILITY is the same
```

## Key Takeaway

- Complex numbers = real part + imaginary part
- Conjugate = flip the sign of the imaginary part
- Modulus squared = probability in quantum computing
- Phase (angle) = invisible in measurement but controls interference
- All qubit amplitudes are complex numbers, and their modulus squared must add to 1
