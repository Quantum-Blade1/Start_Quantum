# 06 - Eigenvalues and Eigenvectors

When you measure a qubit, the result you get is an eigenvalue. The state it collapses to is an eigenvector. This is not a metaphor. This is literally how quantum measurement works mathematically.

If you only understand one thing from this notebook, understand this: eigenvalues = measurement outcomes, eigenvectors = the states after measurement.

```python
import numpy as np
```

## What are Eigenvalues and Eigenvectors?

For a matrix A, if there is a vector v and a scalar lambda such that:

`A @ v = lambda * v`

Then:
- `v` is an eigenvector of A
- `lambda` is the corresponding eigenvalue

In plain English: when you apply the matrix to this special vector, the vector does not change direction. It only gets scaled by the eigenvalue.

```python
# Pauli Z matrix
Z = np.array([[1, 0],
              [0, -1]], dtype=complex)

ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

# Z|0> = 1 * |0>  -->  eigenvalue = +1, eigenvector = |0>
print("Z|0> =")
print(Z @ ket_0)
print("That is +1 * |0>. So |0> is an eigenvector with eigenvalue +1.")

# Z|1> = -1 * |1>  -->  eigenvalue = -1, eigenvector = |1>
print("\nZ|1> =")
print(Z @ ket_1)
print("That is -1 * |1>. So |1> is an eigenvector with eigenvalue -1.")
```

## This IS Quantum Measurement

When you measure a qubit in the Z basis (the standard basis):
- You get result `+1` with some probability, and the qubit collapses to |0>
- You get result `-1` with some probability, and the qubit collapses to |1>

Those `+1` and `-1` are the eigenvalues of Z. The |0> and |1> are the eigenvectors of Z. That is the entire connection.

## Computing Eigenvalues and Eigenvectors with NumPy

```python
# Let NumPy compute them for us
Z = np.array([[1, 0], [0, -1]], dtype=complex)

eigenvalues, eigenvectors = np.linalg.eigh(Z)  # eigh for Hermitian matrices

print("Eigenvalues of Z:", eigenvalues)
print("\nEigenvectors of Z (as columns):")
print(eigenvectors)

# Each column of eigenvectors corresponds to an eigenvalue
for i in range(len(eigenvalues)):
    val = eigenvalues[i]
    vec = eigenvectors[:, i:i+1]
    print(f"\nEigenvalue {val}: eigenvector = {vec.flatten()}")
    # Verify: A @ v = lambda * v
    lhs = Z @ vec
    rhs = val * vec
    print(f"  A @ v = {lhs.flatten()}, lambda * v = {rhs.flatten()}, Match: {np.allclose(lhs, rhs)}")
```

## Eigenvalues of Other Quantum Gates

```python
# Pauli X
X = np.array([[0, 1], [1, 0]], dtype=complex)
vals, vecs = np.linalg.eigh(X)
print("Pauli X:")
print(f"  Eigenvalues: {vals}")
print(f"  Eigenvectors: {vecs[:, 0].round(4)}, {vecs[:, 1].round(4)}")
# The eigenvectors of X are |+> and |-> (the Hadamard basis)

print()

# Hadamard
H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
vals, vecs = np.linalg.eigh(H)
print("Hadamard:")
print(f"  Eigenvalues: {vals.round(4)}")
print(f"  Eigenvectors: {vecs[:, 0].round(4)}, {vecs[:, 1].round(4)}")
```

## Diagonalization

A matrix A can be written as: `A = P @ D @ P^(-1)`

Where:
- D is a diagonal matrix with eigenvalues on the diagonal
- P is a matrix whose columns are the eigenvectors

This is useful because diagonal matrices are easy to work with. Powers of A become easy: `A^n = P @ D^n @ P^(-1)`

```python
# Diagonalize Pauli Z
Z = np.array([[1, 0], [0, -1]], dtype=complex)

eigenvalues, P = np.linalg.eigh(Z)
D = np.diag(eigenvalues)

print("P (eigenvector matrix):")
print(P)
print("\nD (diagonal eigenvalue matrix):")
print(D)

# Reconstruct: A = P @ D @ P^(-1)
reconstructed = P @ D @ np.linalg.inv(P)
print("\nReconstructed Z:")
print(np.round(reconstructed, 10))
print(f"\nMatches original? {np.allclose(Z, reconstructed)}")
```

## Spectral Decomposition

Any Hermitian matrix can be written as a sum of its eigenvalues times the outer products of its eigenvectors:

`A = lambda_1 * |v1><v1| + lambda_2 * |v2><v2| + ...`

This connects eigenvalues back to the outer products we learned in notebook 03.

```python
# Spectral decomposition of Pauli Z
# Z = (+1)|0><0| + (-1)|1><1|

ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

Z_reconstructed = 1 * (ket_0 @ ket_0.conj().T) + (-1) * (ket_1 @ ket_1.conj().T)

print("Z from spectral decomposition:")
print(Z_reconstructed)
print(f"\nMatches Pauli Z? {np.allclose(Z_reconstructed, Z)}")
```

## Key Takeaway

| Concept | Math | Quantum Meaning |
|---------|------|-----------------|
| Eigenvalue | Scalar lambda | Measurement outcome |
| Eigenvector | Vector v | State after measurement |
| Diagonalization | A = P D P^-1 | Changing measurement basis |
| Spectral decomposition | Sum of eigenvalue * projectors | Full description of an observable |
