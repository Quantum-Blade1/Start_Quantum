# Eigenvalues and Eigenvectors: The Measurement Foundation

Eigenvalues and eigenvectors are **the heart of quantum measurement**. They encode what outcomes you can measure and what states give those outcomes with certainty.

## Definition: Eigenvalue Equation

For a matrix A, an **eigenvector** |v⟩ and **eigenvalue** λ satisfy:
$$A|v\rangle = \lambda |v\rangle$$

Reading this equation: "Applying A to |v⟩ just scales it by λ; the direction doesn't change."

**Key insight**: If |ψ⟩ is an eigenvector of observable H with eigenvalue λ, then:
- Measuring H on state |ψ⟩ **always gives outcome λ** (with probability 1)
- The state **doesn't collapse**; it remains |ψ⟩

This is why eigenstates are also called **measurement eigenstates** or **eigenspaces**.

## Computing Eigenvalues: The Characteristic Polynomial

To find eigenvalues, solve det(A - λI) = 0.

For the Pauli-Z matrix:
$$\sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

$$\det(\sigma_z - \lambda I) = \det\begin{pmatrix} 1-\lambda & 0 \\ 0 & -1-\lambda \end{pmatrix} = (1-\lambda)(-1-\lambda) = \lambda^2 - 1 = 0$$

So λ = ±1. The eigenvalues are the measurement outcomes!

For the Hadamard gate:
$$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$$

$$\det(H - \lambda I) = \frac{1}{2}[(1-\sqrt{2}\lambda)(-1-\sqrt{2}\lambda)] = \lambda^2 - 1$$

Again λ = ±1. All unitary matrices have unit-modulus eigenvalues.

## Finding Eigenvectors: Null Space of A - λI

For each eigenvalue λ, solve (A - λI)|v⟩ = 0.

For σz with λ = 1:
$$(\sigma_z - I)|v\rangle = \begin{pmatrix} 0 & 0 \\ 0 & -2 \end{pmatrix}\begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This gives v₂ = 0, so |v⟩ = |0⟩ (up to normalization).

For σz with λ = -1:
$$(\sigma_z + I)|v\rangle = \begin{pmatrix} 2 & 0 \\ 0 & 0 \end{pmatrix}\begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$

This gives v₁ = 0, so |v⟩ = |1⟩.

So σz has eigenstates |0⟩, |1⟩ with eigenvalues 1, -1 respectively. These are the computational basis states!

## Spectral Decomposition: Diagonalizing Observables

For Hermitian H with eigenvalues {λᵢ} and orthonormal eigenvectors {|uᵢ⟩}:

$$H = \sum_i \lambda_i |u_i\rangle\langle u_i|$$

This **spectral decomposition** is powerful:
1. **Measurement probabilities**: P(λᵢ) = |⟨uᵢ|ψ⟩|²
2. **Expectation value**: ⟨H⟩ = Σᵢ λᵢ P(λᵢ)
3. **Functions of H**: eᴴ, e^{-iHt} all defined via spectral decomposition

For example, e^{-iσzt} (time evolution under σz):
$$e^{-i\sigma_z t} = e^{-it} |0\rangle\langle 0| + e^{it} |1\rangle\langle 1|$$

## Change of Basis: Diagonalization Matrix

If P has columns = eigenvectors of A, and Λ = diag(λ₁, λ₂,...), then:
$$A = P \Lambda P^{-1}$$

For Hermitian matrices, P is unitary (P† = P⁻¹), so:
$$A = P \Lambda P^\dagger$$

**Physically**: The matrix P represents a **change of measurement basis**. The diagonal Λ shows measurement outcomes.

## Degeneracy: Multiple Eigenvectors per Eigenvalue

When multiple eigenvectors share the same eigenvalue, the eigenvalue is **degenerate**.

Example: Identity I has eigenvalue 1 with multiplicity 2. ANY vector is an eigenvector!
$$I|v\rangle = |v\rangle \text{ for all } |v\rangle$$

This degeneracy means "you can't uniquely determine the state by this measurement." Degenerate measurements reveal less information.

## Generalized Eigenvalue Problem

Sometimes we solve Av = λBv (not just Av = λv).

In quantum computing, this appears in:
- **Generalized eigenvalue problems** in Hermitian operators: Hv = λSv
- **Variational Eigensolvers** (VQE): minimize smallest eigenvalue iteratively

## Numerical Considerations

For large matrices (thousands of qubits = 2^n × 2^n matrices!), eigenvalue computation becomes numerically challenging:
- Full diagonalization O(n³) complexity infeasible for n > 10
- Iterative methods (Lanczos, ARPACK) compute k smallest eigenvalues efficiently
- This limitation motivates variational quantum algorithms!

```python
# Functional code: Eigenvalues and eigenvector decomposition
import numpy as np

# Define observables
sigma_z = np.array([[1, 0], [0, -1]])
sigma_x = np.array([[0, 1], [1, 0]])
H_gate = np.array([[1, 1], [1, -1]]) / np.sqrt(2)

# Compute eigenvalues and eigenvectors
evals_z, evecs_z = np.linalg.eigh(sigma_z)  # eigenvalues, eigenvectors (columns)
evals_x, evecs_x = np.linalg.eigh(sigma_x)

# Characteristic polynomial roots (alternative method)
char_poly_coeffs = np.poly(sigma_z)  # coefficients of det(A - λI)

# Verify eigenvalue equation: A|v⟩ = λ|v⟩
v0_z = evecs_z[:, 0]  # First eigenvector
Av = sigma_z @ v0_z
lambda_v = evals_z[0] * v0_z
is_eigenvalue = np.allclose(Av, lambda_v)

# Spectral decomposition: A = ΣᵢλᵢPᵢ where Pᵢ = |uᵢ⟩⟨uᵢ|
P0 = np.outer(evecs_z[:, 0], evecs_z[:, 0].conj())  # |0⟩⟨0|
P1 = np.outer(evecs_z[:, 1], evecs_z[:, 1].conj())  # |1⟩⟨1|
sigma_z_reconstructed = evals_z[0] * P0 + evals_z[1] * P1

# Measurement probabilities: P(λᵢ) = |⟨uᵢ|ψ⟩|²
psi = (np.array([1, 0]) + np.array([0, 1])) / np.sqrt(2)  # |+⟩
prob_0 = np.abs(np.vdot(evecs_z[:, 0], psi))**2  # P(λ₀)
prob_1 = np.abs(np.vdot(evecs_z[:, 1], psi))**2  # P(λ₁)

# Expectation value: ⟨H⟩ = Σᵢ λᵢ P(λᵢ)
expectation_z = np.sum(evals_z * np.array([prob_0, prob_1]))

# Time evolution: e^{-iHt}
t = 0.5
U_t = np.diag(np.exp(-1j * evals_z * t))  # Diagonal in eigenbasis
# Transform back: e^{-iHt} = P·e^{-iΛt}·P†
time_evolution = evecs_z @ U_t @ evecs_z.conj().T

# Diagonalization: A = PΛP^{-1}
Lambda = np.diag(evals_z)
A_reconstructed = evecs_z @ Lambda @ np.linalg.inv(evecs_z)

# Functions of matrices: e^H via spectral decomposition
exp_sigma_z = evecs_z @ np.diag(np.exp(evals_z)) @ evecs_z.conj().T

# Degeneracy check: Identity matrix
identity = np.eye(2)
evals_I, evecs_I = np.linalg.eigh(identity)  # Should all be λ = 1
```

## Summary: Eigenvectors = Measurement Eigenstates

✓ **Eigenvalue equation**: A|v⟩ = λ|v⟩; λ are measurement outcomes  
✓ **Characteristic polynomial**: det(A - λI) = 0 finds eigenvalues  
✓ **Spectral decomposition**: H = Σᵢ λᵢ|uᵢ⟩⟨uᵢ| encodes all measurement physics  
✓ **Measurement probability**: P(λᵢ) = |⟨uᵢ|ψ⟩|² from projectors onto eigenstates  
✓ **Expectation value**: ⟨H⟩ = Πᵢ λᵢ P(λᵢ) = weighted average of eigenvalues  
✓ **Time evolution**: e^{-iHt} = P·e^{-iΛt}·P† via spectral decomposition  
✓ **Degeneracy**: Multiple eigenvectors per eigenvalue = ambiguous measurement outcomes  

Eigenvector analysis is how quantum computers extract classical information from quantum states.
