In German, "Eigen" means "own" or "characteristic". These are the characteristic finger-prints of a matrix.

---

## 1. The Eigenvalue Equation

Let $A$ be a square matrix.
Usually, when $A$ acts on a vector $|v\rangle$, it rotates it and stretches it. The new vector points in a different direction.

But for some special vectors, applying $A$ *does not change their direction*. It only stretches or shrinks them.

$$ A|v\rangle = \lambda|v\rangle $$

- $|v\rangle$ is an **Eigenvector**.
- $\lambda$ is the corresponding **Eigenvalue** (a scalar).

This equation is the heart of quantum mechanics.

---

## 2. The Measurement Axiom

Here is how physics maps to math:

1. **Observable:** A physical quantity (like Energy or Spin) is represented by a **Hermitian Matrix** $H$.
2. **Possible Outcomes:** The only values you can ever measure are the **Eigenvalues** of $H$.
3. **Post-Measurement State:** If you measure eigenvalue $\lambda$, the system instantly collapses into the corresponding **Eigenvector** $|v\rangle$.

### Example: The Z Gate

The Z gate is also an observable (Spin-Z).

$$ Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} $$

Let's check its action on $|0\rangle$:
$$ Z|0\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \end{pmatrix} = 1|0\rangle $$
So $|0\rangle$ is an eigenvector with eigenvalue **+1**.

Let's check its action on $|1\rangle$:
$$ Z|1\rangle = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 0 \\ -1 \end{pmatrix} = -1|1\rangle $$
So $|1\rangle$ is an eigenvector with eigenvalue **-1**.

Conclusion: If you measure Spin-Z, you can only get +1 (spin up) or -1 (spin down). You can never get 0.5 or 0.

```python
import numpy as np

# Define Z matrix
Z = np.array([[1, 0], 
              [0, -1]])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(Z)

print("Eigenvalues of Z:")
print(eigenvalues)

print("\nEigenvectors of Z:")
# Note: eigenvectors are columns
for i in range(len(eigenvalues)):
    vec = eigenvectors[:, i]
    print(f"Vector {i}: {vec}")
```

## 3. The X Basis

What about the X gate?

$$ X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} $$

Does it have eigenvectors? Yes.

$$ X|+\rangle = |+\rangle $$
$$ X|-\rangle = -|-\rangle $$

Wait, let's verify that manually.
$|+\rangle = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ 1 \end{pmatrix}$

$$ X|+\rangle = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = |+\rangle $$

So $|+\rangle$ is the eigenvector with eigenvalue +1.
And $|-\rangle$ is the eigenvector with eigenvalue -1.

This means if you measure Spin-X, the possible states are $|+\rangle$ and $|-\rangle$.

```python
X = np.array([[0, 1], 
              [1, 0]])

evals_X, evecs_X = np.linalg.eig(X)

print("Eigenvalues of X:")
print(evals_X)

print("\nEigenvectors of X:")
print(evecs_X)

# Notice [0.707, 0.707] is |+>
# Notice [-0.707, 0.707] is proportional to |->
```

## 4. Spectral Decomposition

Because the eigenvectors of a Hermitian matrix form an orthonormal basis, we can write the matrix as a sum of projectors.

$$ A = \sum_i \lambda_i |v_i\rangle \langle v_i| $$

For measuring Z:
$$ Z = (+1)|0\rangle\langle 0| + (-1)|1\rangle\langle 1| $$
$$ = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} - \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} $$

This is why "diagonalizing a matrix" is the same as finding the basis where measurements make sense.
