# 02 - Vectors and Dirac Notation

A qubit is a vector. That is the single most important sentence in quantum computing.

The state |0> is a vector. The state |1> is a vector. Any qubit state |psi> is a vector. Quantum gates take a vector as input and return a vector as output. If you understand vectors, you already understand half of quantum computing.

## Vectors as Arrays

A vector is just an ordered list of numbers. In quantum computing, we use column vectors.

- |0> = [1, 0] (first element is 1, second is 0)
- |1> = [0, 1] (first element is 0, second is 1)

These two vectors form the computational basis. Every single-qubit state can be written as a combination of these two.

```python
import numpy as np

# The computational basis states
ket_0 = np.array([[1], [0]], dtype=complex)  # |0>
ket_1 = np.array([[0], [1]], dtype=complex)  # |1>

print("|0> =")
print(ket_0)
print("\n|1> =")
print(ket_1)
```

## Dirac Notation (Bra-Ket)

Quantum computing uses a special notation invented by Paul Dirac:

- **Ket** `|a>` = column vector (the state of the qubit)
- **Bra** `<a|` = row vector (the conjugate transpose of the ket)

To go from ket to bra: transpose the vector AND take the complex conjugate of each element.

This notation looks weird at first but it makes equations much cleaner once you get used to it.

```python
# Ket |0> is a column vector
ket_0 = np.array([[1], [0]], dtype=complex)
print("|0> (ket) =")
print(ket_0)

# Bra <0| is the conjugate transpose (dagger) of the ket
bra_0 = ket_0.conj().T
print("\n<0| (bra) =")
print(bra_0)

# With complex amplitudes it matters more
ket_psi = np.array([[1/np.sqrt(2)], [1j/np.sqrt(2)]], dtype=complex)
bra_psi = ket_psi.conj().T

print("\n|psi> =")
print(ket_psi)
print("\n<psi| =")
print(bra_psi)
# Notice: the 'j' became '-j' because of the conjugate
```

## Common Qubit States

These states come up all the time. Memorize them.

| State | Vector | Description |
|-------|--------|-------------|
| \|0> | [1, 0] | Basis state zero |
| \|1> | [0, 1] | Basis state one |
| \|+> | [1/sqrt(2), 1/sqrt(2)] | Equal superposition (same phase) |
| \|-> | [1/sqrt(2), -1/sqrt(2)] | Equal superposition (opposite phase) |

```python
# All four common states
states = {
    "|0>": np.array([[1], [0]], dtype=complex),
    "|1>": np.array([[0], [1]], dtype=complex),
    "|+>": np.array([[1/np.sqrt(2)], [1/np.sqrt(2)]], dtype=complex),
    "|->": np.array([[1/np.sqrt(2)], [-1/np.sqrt(2)]], dtype=complex),
}

for name, state in states.items():
    p0 = abs(state[0, 0])**2
    p1 = abs(state[1, 0])**2
    print(f"{name}: P(0)={p0:.4f}, P(1)={p1:.4f}")
```

## Superposition: Linear Combination of Vectors

Any qubit state can be written as:

`|psi> = alpha * |0> + beta * |1>`

This is just vector addition with scalar multiplication. Nothing more.

```python
ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

# Build |+> from |0> and |1>
alpha = 1/np.sqrt(2)
beta = 1/np.sqrt(2)
ket_plus = alpha * ket_0 + beta * ket_1

print("|+> built from linear combination:")
print(ket_plus)

# Build a custom state: 60% chance of |0>, 40% chance of |1>
alpha = np.sqrt(0.6)
beta = np.sqrt(0.4)
custom_state = alpha * ket_0 + beta * ket_1

print(f"\nCustom state: {alpha:.4f}|0> + {beta:.4f}|1>")
print(f"P(0) = {abs(alpha)**2:.4f}")
print(f"P(1) = {abs(beta)**2:.4f}")
print(f"Total = {abs(alpha)**2 + abs(beta)**2:.4f}")
```

## Normalization

A valid qubit state must be normalized. This means the sum of the squared magnitudes of all amplitudes must equal 1:

`|alpha|^2 + |beta|^2 = 1`

If your vector does not satisfy this, it is not a valid quantum state. You can normalize any vector by dividing it by its length (norm).

```python
# This is NOT a valid quantum state (not normalized)
bad_state = np.array([[3], [4]], dtype=complex)
norm = np.linalg.norm(bad_state)
print(f"Bad state norm: {norm}")  # Not 1

# Normalize it by dividing by the norm
good_state = bad_state / norm
print(f"\nNormalized state:")
print(good_state)
print(f"Norm after normalization: {np.linalg.norm(good_state):.4f}")  # Now it is 1

# Check probabilities
p0 = abs(good_state[0, 0])**2
p1 = abs(good_state[1, 0])**2
print(f"P(0) = {p0:.4f}, P(1) = {p1:.4f}, Total = {p0+p1:.4f}")
```

## Vector Operations Summary

| Operation | What it does | Quantum meaning |
|-----------|-------------|----------------|
| Scalar multiplication | Scale a vector | Change amplitude |
| Vector addition | Combine vectors | Superposition |
| Conjugate transpose | Ket to Bra | Needed for inner products |
| Normalization | Make length = 1 | Valid quantum state |
