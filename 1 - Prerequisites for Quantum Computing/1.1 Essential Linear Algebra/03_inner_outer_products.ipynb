# 03 - Inner Product and Outer Product

The inner product tells you how similar two quantum states are. The outer product builds operators that act on quantum states. Both come up constantly in quantum computing.

## Inner Product (Bracket)

The inner product of two states |a> and |b> is written as `<a|b>` (bra-ket, or bracket -- that is literally where the name comes from).

To compute it: take the conjugate transpose of |a>, then multiply by |b>.

`<a|b> = (conjugate transpose of a) * b`

The result is a single complex number (a scalar, not a vector).

```python
import numpy as np

# Basis states
ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

# Inner product <0|0> -- same state with itself
inner_00 = ket_0.conj().T @ ket_0
print(f"<0|0> = {inner_00[0,0]}")  # 1 (perfectly overlapping)

# Inner product <1|1>
inner_11 = ket_1.conj().T @ ket_1
print(f"<1|1> = {inner_11[0,0]}")  # 1

# Inner product <0|1> -- different orthogonal states
inner_01 = ket_0.conj().T @ ket_1
print(f"<0|1> = {inner_01[0,0]}")  # 0 (no overlap at all)

# Inner product <1|0>
inner_10 = ket_1.conj().T @ ket_0
print(f"<1|0> = {inner_10[0,0]}")  # 0
```

## What the Inner Product Means Physically

- `<a|b> = 1` means the states are identical
- `<a|b> = 0` means the states are orthogonal (completely distinguishable)
- `|<a|b>|^2` is the probability of measuring state |a> when the system is in state |b>

This is how measurement works in quantum computing. The probability of getting a specific result is the inner product squared.

```python
# State |+> = (|0> + |1>) / sqrt(2)
ket_plus = np.array([[1/np.sqrt(2)], [1/np.sqrt(2)]], dtype=complex)

# What is the probability of measuring |0> when the qubit is in state |+>?
amplitude = ket_0.conj().T @ ket_plus   # <0|+>
probability = abs(amplitude[0,0])**2
print(f"<0|+> = {amplitude[0,0]:.4f}")
print(f"|<0|+>|^2 = {probability:.4f}")  # 0.5 -- 50% chance

# What about measuring |1>?
amplitude = ket_1.conj().T @ ket_plus   # <1|+>
probability = abs(amplitude[0,0])**2
print(f"\n<1|+> = {amplitude[0,0]:.4f}")
print(f"|<1|+>|^2 = {probability:.4f}")  # 0.5 -- 50% chance
```

## Orthogonality and Orthonormal Bases

Two vectors are **orthogonal** if their inner product is 0. They are **orthonormal** if they are orthogonal AND each has a norm of 1.

The computational basis {|0>, |1>} is orthonormal:
- <0|0> = 1, <1|1> = 1 (normalized)
- <0|1> = 0, <1|0> = 0 (orthogonal)

The Hadamard basis {|+>, |->} is also orthonormal.

```python
ket_plus = np.array([[1/np.sqrt(2)], [1/np.sqrt(2)]], dtype=complex)
ket_minus = np.array([[1/np.sqrt(2)], [-1/np.sqrt(2)]], dtype=complex)

# Check orthonormality of {|+>, |->}
print(f"<+|+> = {(ket_plus.conj().T @ ket_plus)[0,0]:.4f}")   # 1
print(f"<-|-> = {(ket_minus.conj().T @ ket_minus)[0,0]:.4f}") # 1
print(f"<+|-> = {(ket_plus.conj().T @ ket_minus)[0,0]:.4f}")  # 0
print(f"<-|+> = {(ket_minus.conj().T @ ket_plus)[0,0]:.4f}")  # 0
print("\n{|+>, |->} is an orthonormal basis.")
```

## Outer Product

The outer product of |a> and <b| is written as `|a><b|`.

Unlike the inner product (which gives a scalar), the outer product gives a **matrix**.

`|a><b| = column_vector * row_vector = matrix`

Outer products are used to build projection operators and density matrices.

```python
ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

# Outer product |0><0| -- projects onto state |0>
P0 = ket_0 @ ket_0.conj().T
print("|0><0| =")
print(P0)

# Outer product |1><1| -- projects onto state |1>
P1 = ket_1 @ ket_1.conj().T
print("\n|1><1| =")
print(P1)

# These are projection operators
# P0 + P1 should equal the identity matrix (completeness)
print("\n|0><0| + |1><1| =")
print(P0 + P1)  # Identity matrix
```

## Projection: What Happens When You Measure

When you measure a qubit, it gets projected onto one of the basis states. The projection operator `|0><0|` applied to a state gives you the component of that state in the |0> direction.

```python
# State |+>
ket_plus = np.array([[1/np.sqrt(2)], [1/np.sqrt(2)]], dtype=complex)

# Project |+> onto |0>
P0 = ket_0 @ ket_0.conj().T
projected = P0 @ ket_plus
print("P0 |+> =")
print(projected)

# The probability of measuring |0> is the norm squared of the projected state
prob = np.linalg.norm(projected)**2
print(f"\nProbability of measuring |0>: {prob:.4f}")

# Project |+> onto |1>
P1 = ket_1 @ ket_1.conj().T
projected = P1 @ ket_plus
prob = np.linalg.norm(projected)**2
print(f"Probability of measuring |1>: {prob:.4f}")
```

## Quick Reference

| Operation | Notation | Input | Output | Quantum Use |
|-----------|----------|-------|--------|-------------|
| Inner product | <a\|b> | Two vectors | Scalar | Measurement probability |
| Outer product | \|a><b\| | Two vectors | Matrix | Projection operators |
