# Chapter 05: Matrices and Linear Transformations — The Actions

Welcome to **Section 1.1: Essential Linear Algebra**.

We have states (vectors) and a language to describe them (Dirac notation). Now we need to **do something** with them.

In quantum computing, every operation you perform on a qubit — every logic gate, every time step, every evolution — is a **linear transformation**.

And linear transformations are represented by **Matrices**.

If $|\psi\rangle$ is the state of a qubit, applying a gate $U$ transforms it into a new state $|\psi'\rangle$:

$$ |\psi'\rangle = U|\psi\rangle $$

This notebook will explain matrices not as grids of numbers, but as **machines** that transform space.

---

## 1. Linearity: The Golden Rule

A transformation $A$ is **linear** if it preserves two things:

1. **Additivity:** $A(|u\rangle + |v\rangle) = A|u\rangle + A|v\rangle$
2. **Homogeneity:** $A(c|v\rangle) = c(A|v\rangle)$

Combine them:
$$ A(\alpha|u\rangle + \beta|v\rangle) = \alpha A|u\rangle + \beta A|v\rangle $$

### Why is Quantum Mechanics Linear?

This is a deep fact about nature. It means if a gate flips $|0\rangle$ to $|1\rangle$ ($X|0\rangle = |1\rangle$), and flips $|1\rangle$ to $|0\rangle$ ($X|1\rangle = |0\rangle$)...

...then automatically, we know what it does to a superposition:
$$ X(\frac{|0\rangle + |1\rangle}{\sqrt{2}}) = \frac{X|0\rangle + X|1\rangle}{\sqrt{2}} = \frac{|1\rangle + |0\rangle}{\sqrt{2}} $$

If quantum mechanics were non-linear, we could solve NP-complete problems instantly (and maybe time travel). But as far as we know, it is strictly linear.

---

## 2. Matrices as Operators

A matrix is just a way to write down a linear transformation.

$$ \begin{pmatrix} a & b \\ c & d \end{pmatrix} $$

What do these numbers mean?

The columns tell you **where the basis vectors go**.

- Column 1: $\begin{pmatrix} a \\ c \end{pmatrix}$ is the new state of $|0\rangle$.
- Column 2: $\begin{pmatrix} b \\ d \end{pmatrix}$ is the new state of $|1\rangle$.

Check it:
$$ \begin{pmatrix} a & b \\ c & d \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} a \\ c \end{pmatrix} $$
$$ \begin{pmatrix} a & b \\ c & d \end{pmatrix} \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} b \\ d \end{pmatrix} $$

This is a powerful intuition. If you want to construct a matrix that swaps $|0\rangle$ and $|1\rangle$, you just put $|1\rangle$ in the first column and $|0\rangle$ in the second column.

$$ X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} $$

```python
import numpy as np

# Define the X gate matrix
X = np.array([[0, 1], 
              [1, 0]])

# Define basis states
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

# Apply X gate (Matrix multiplication using @)
output_0 = X @ ket_0
output_1 = X @ ket_1

print(f"X|0> = {output_0} (Should be |1>)")
print(f"X|1> = {output_1} (Should be |0>)")
```

---

## 3. The Hadamard Gate (H)

The most famous quantum gate.

It takes $|0\rangle$ to $|+\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}}$.
It takes $|1\rangle$ to $|-\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$.

So, remembering our "columns are destinations" rule:
- Col 1: $[1/\sqrt{2}, 1/\sqrt{2}]^T$
- Col 2: $[1/\sqrt{2}, -1/\sqrt{2}]^T$

$$ H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix} $$

This gate creates **Superposition**.

```python
# Define H gate
H = (1/np.sqrt(2)) * np.array([[1, 1],
                               [1, -1]])

# Apply to |0>
psi_plus = H @ ket_0

print(f"H|0> = {psi_plus}")
# Verify it is normalized
print(f"Norm: {np.linalg.norm(psi_plus)}")
```

---

## 4. Identity Matrix (I)

The identity matrix does nothing. It leaves the vector unchanged.

$$ I = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} $$

$$ I|\psi\rangle = |\psi\rangle $$

It seems useless, but it is critically important for:
1. Defining inverse matrices ($A A^{-1} = I$)
2. Describing "do nothing" time steps (Memory)
3. Building multi-qubit gates (applying X to qubit 1 and I to qubit 2).

---

## 5. Inverse Matrix

Sometimes we want to undo an operation. If we applied gate $U$, we want to apply gate $U^{-1}$ to get back to where we started.

Every quantum gate (except measurement) is **Reversible**.

This means every valid quantum gate matrix has an inverse. Information is never lost in a closed quantum system. This is different from classical logic: The AND gate is irreversible (if output is 0, you don't know if inputs were 00, 01, or 10).

For the X gate, it is its own inverse ($X X = I$). Flipping a bit twice gives you the original bit.
For the H gate, it is also its own inverse ($H H = I$). Applying H twice gives you back the basis state.

```python
# Verifying H is its own inverse
# @ is matrix multiplication
H_squared = H @ H

print("H * H =")
print(np.round(H_squared)) # Round to remove tiny floating point errors
# Should see Identity matrix
```
