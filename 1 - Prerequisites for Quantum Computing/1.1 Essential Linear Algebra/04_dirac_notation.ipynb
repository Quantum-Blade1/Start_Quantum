# Chapter 04: Dirac Notation — The Language of Quantum Mechanics

Welcome to **Section 1.1: Essential Linear Algebra**.

Physics is full of notation. Vectors have arrows ($\vec{v}$). Matrices have bold letters ($\mathbf{M}$). Integrals have squiggly lines ($\int$).

But in quantum mechanics, one notation rules them all: **Dirac Notation** (also called Bra-Ket notation).

Invented by the legendary physicist Paul Dirac, this notation is not just a shorthand. It is a powerful tool. It abstracts away the tedious details of matrices and integrals and lets you focus on the algebra of states.

If you want to read any paper or book on quantum computing, you **must** be fluent in Dirac notation. It is the alphabet of our field.

In this notebook, we master it.

---

## 1. The Ket: $|\psi\rangle$

A **Ket** is simply a column vector. It represents a quantum state.

$$ |\psi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix} $$

The symbol $|\psi\rangle$ is just a name. You can call it $|v\rangle$, $|x\rangle$, $|alice\rangle$, $|shrodingers\_cat\rangle$. Anything.

Standard basis kets:
- $|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$ (State 0)
- $|1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$ (State 1)

Superposition:
- $|+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ 1 \end{pmatrix}$
- $|-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle) = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 \\ -1 \end{pmatrix}$

Think of a Ket as a "Key" that unlocks a state definition.

---

## 2. The Bra: $\langle\psi|$

A **Bra** is a row vector. But not just any row vector.

It is the **Hermitian conjugate** (conjugate transpose) of the corresponding Ket.

If $|\psi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$, then:

$$ \langle \psi | = (|\psi\rangle)^\dagger = (\alpha^*, \beta^*) $$

Notice:
1. It is horizontal (Row).
2. The elements are complex conjugates (Star).

Why "Bra" and "Ket"? Because when you put them together, they form a "Bra-c-Ket" (bracket).

$$ \langle \phi | \psi \rangle $$

```python
import numpy as np

# Ket |psi> = [i, 2]
ket_psi = np.array([1j, 2])

# Bra <psi| = [-i, 2] (Conjugate Transpose)
bra_psi = ket_psi.conj()

print(f"Ket: {ket_psi}")
print(f"Bra: {bra_psi}")

# Note: Numpy prints 1D arrays horizontally, but mathematically:
# Ket is Column.
# Bra is Row.
```

## 3. The Inner Product (Bra-Ket)

The inner product of $|\phi\rangle$ and $|\psi\rangle$ is written as:

$$ \langle \phi | \psi \rangle $$

This is a **scalar** (a single complex number). It represents the **overlap** between the two states.

$$ \langle \phi | \psi \rangle = \phi_1^* \psi_1 + \phi_2^* \psi_2 + ... $$

Properties in Dirac notation:
- $\langle \psi | \psi \rangle = 1$ (Normalization, if normalized)
- $\langle \psi | \phi \rangle = \langle \phi | \psi \rangle^*$ (Conjugate symmetry)
- $\langle \phi | (a|\psi\rangle + b|\chi\rangle) = a\langle \phi | \psi \rangle + b\langle \phi | \chi \rangle$ (Linearity in Ket)

### Example: Calculating <0|+>

We know $|+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$.

$$ \langle 0 | + \rangle = \langle 0 | (\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle) $$
$$ = \frac{1}{\sqrt{2}} \langle 0 | 0 \rangle + \frac{1}{\sqrt{2}} \langle 0 | 1 \rangle $$

Since $\langle 0 | 0 \rangle = 1$ and $\langle 0 | 1 \rangle = 0$:

$$ = \frac{1}{\sqrt{2}}(1) + 0 $$
$$ = \frac{1}{\sqrt{2}} $$

This tells us the amplitude of finding $|0\rangle$ in the state $|+\rangle$ is $1/\sqrt{2}$. The probability is $|1/\sqrt{2}|^2 = 1/2$.

```python
# Inner Product: <psi|psi>
# Manually: (1j)*(-1j) + (2)*(2) = 1 + 4 = 5
norm_squared = np.vdot(ket_psi, ket_psi)

print(f"<psi|psi> (Norm Squared): {norm_squared}")

# Normalization
norm = np.sqrt(norm_squared.real)
ket_psi_norm = ket_psi / norm
print(f"Normalized Ket: {ket_psi_norm}")
print(f"Check norm: {np.vdot(ket_psi_norm, ket_psi_norm).real}") # Should be 1.0
```

## 4. The Outer Product (Ket-Bra)

What happens if you reverse the order?

$$ |\psi\rangle \langle \phi | $$

This is NOT a scalar. This is a **Matrix** (Operator).

Column $\times$ Row = Matrix.

$$ \begin{pmatrix} \alpha \\ \beta \end{pmatrix} \begin{pmatrix} \gamma^* & \delta^* \end{pmatrix} = \begin{pmatrix} \alpha\gamma^* & \alpha\delta^* \\ \beta\gamma^* & \beta\delta^* \end{pmatrix} $$

### Why Outer Products Matter

Outer products allow us to build operators (matrices) from vectors.

1. **Projection Operator:** $P = |\psi\rangle\langle\psi|$. This operator projects any state onto the direction of $|\psi\rangle$.
   Example: $P_0 = |0\rangle\langle 0| = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$.
   It kills the $|1\rangle$ component and keeps the $|0\rangle$ component.

2. **Representation of Operators:** ANY operator $A$ can be written as sum of outer products.
   $A = \sum_i \lambda_i |v_i\rangle\langle v_i|$ (Spectral Decomposition).
   This is deep. It means if you know the eigenvectors ($|v_i\rangle$) and eigenvalues ($\lambda_i$), you know the matrix.

```python
# Outer Product: |0><0|
ket_0 = np.array([1, 0])
proj_0 = np.outer(ket_0, ket_0.conj())

print("Projection Operator |0><0|:")
print(proj_0)

# Apply |0><0| to |+>
# Expect: 1/sqrt(2) |0>
ket_plus = np.array([1, 1]) / np.sqrt(2)
result = proj_0 @ ket_plus

print(f"|+>: {ket_plus}")
print(f"Projected |+>: {result}")
# Note the second component became 0.
```

## 5. Dual Correspondence

This is a pro tip for checking your math.

Every valid equation in Dirac notation has a "dual" equation. You get the dual by:
1. Changing every bra to a ket ($|\psi\rangle \to \langle\psi|$)
2. Changing every ket to a bra ($|\psi\rangle \to \langle\psi|$)
3. Taking the complex conjugate of all scalars ($c \to c^*$)
4. Taking the adjoint of all operators ($A \to A^\dagger$)
5. Reversing the order of multiplication.

Example:
$$ \langle \phi | A | \psi \rangle = c $$
Dual:
$$ \langle \psi | A^\dagger | \phi \rangle = c^* $$

This symmetry is beautiful and useful.
