# 05 - Special Matrices in Quantum Computing

Not every matrix is a valid quantum gate. Only unitary matrices qualify. And the matrices you measure with are Hermitian. This notebook covers the matrix types that actually show up in quantum computing and why they matter.

```python
import numpy as np

# Helper: check if two matrices are (approximately) equal
def is_close(A, B):
    return np.allclose(A, B)
```

## Hermitian Matrices (Observables)

A matrix is Hermitian if it equals its own conjugate transpose: `A = A^dagger`

**Why it matters:** In quantum mechanics, things you can measure (like energy, spin, position) are represented by Hermitian matrices. These are called observables. The measurement outcomes are the eigenvalues of these matrices.

```python
# Pauli Z matrix -- Hermitian
Z = np.array([[1, 0],
              [0, -1]], dtype=complex)

print("Z =")
print(Z)
print("\nZ^dagger =")
print(Z.conj().T)
print(f"\nIs Z Hermitian? {is_close(Z, Z.conj().T)}")

# A non-Hermitian matrix
A = np.array([[1, 1j],
              [1j, 1]], dtype=complex)
print(f"\nIs A Hermitian? {is_close(A, A.conj().T)}")
# Note: for Hermitian, off-diagonal elements must be conjugates of each other
```

## Unitary Matrices (Quantum Gates)

A matrix U is unitary if: `U @ U^dagger = U^dagger @ U = I` (identity)

**Why it matters:** Every quantum gate must be unitary. This is a fundamental rule of quantum mechanics. It guarantees:
- The operation is reversible (you can undo any gate)
- Probabilities are preserved (they still add up to 1 after the gate)

```python
# Hadamard gate
H = np.array([[1, 1],
              [1, -1]], dtype=complex) / np.sqrt(2)

# Check: H @ H^dagger should be identity
product = H @ H.conj().T
print("H @ H^dagger =")
print(np.round(product, 10))
print(f"\nIs H unitary? {is_close(product, np.eye(2))}")

# Pauli X gate
X = np.array([[0, 1],
              [1, 0]], dtype=complex)
print(f"\nIs X unitary? {is_close(X @ X.conj().T, np.eye(2))}")
```

## The Pauli Matrices

The three Pauli matrices (plus the identity) form a basis for all 2x2 matrices. They are the most important matrices in quantum computing.

- **I** (Identity): Does nothing
- **X** (Pauli-X): Bit flip (like classical NOT)
- **Y** (Pauli-Y): Bit flip with a phase
- **Z** (Pauli-Z): Phase flip (flips the sign of |1>)

All Pauli matrices are both Hermitian AND unitary.

```python
# The four Pauli matrices
I = np.array([[1, 0], [0, 1]], dtype=complex)
X = np.array([[0, 1], [1, 0]], dtype=complex)
Y = np.array([[0, -1j], [1j, 0]], dtype=complex)
Z = np.array([[1, 0], [0, -1]], dtype=complex)

paulis = {'I': I, 'X': X, 'Y': Y, 'Z': Z}

ket_0 = np.array([[1], [0]], dtype=complex)
ket_1 = np.array([[0], [1]], dtype=complex)

print("What each Pauli does to |0> and |1>:\n")
for name, P in paulis.items():
    r0 = P @ ket_0
    r1 = P @ ket_1
    print(f"{name}|0> = [{r0[0,0]:.1f}, {r0[1,0]:.1f}]")
    print(f"{name}|1> = [{r1[0,0]:.1f}, {r1[1,0]:.1f}]")
    print()
```

```python
# Verify: all Paulis are both Hermitian and Unitary
for name, P in paulis.items():
    hermitian = is_close(P, P.conj().T)
    unitary = is_close(P @ P.conj().T, np.eye(2))
    print(f"{name}: Hermitian={hermitian}, Unitary={unitary}")
```

## The Hadamard Gate

The Hadamard gate creates superposition. It is used at the start of almost every quantum algorithm.

- H|0> = |+> (equal superposition)
- H|1> = |-> (equal superposition with opposite phase)
- H applied twice = Identity (it undoes itself)

```python
H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
ket_0 = np.array([[1], [0]], dtype=complex)

# H|0> = |+>
print("H|0> =")
print(H @ ket_0)

# H applied twice = I
print("\nH @ H =")
print(np.round(H @ H, 10))
print(f"\nH @ H = I? {is_close(H @ H, np.eye(2))}")
```

## Phase Gates (S and T)

These gates do not change probabilities. They only change the phase of the |1> component. Phase matters for interference in quantum algorithms.

```python
# S gate: adds a phase of pi/2 (i) to |1>
S = np.array([[1, 0],
              [0, 1j]], dtype=complex)

# T gate: adds a phase of pi/4 to |1>
T = np.array([[1, 0],
              [0, np.exp(1j * np.pi/4)]], dtype=complex)

ket_plus = np.array([[1], [1]], dtype=complex) / np.sqrt(2)

# Apply S to |+>
after_S = S @ ket_plus
print("S|+> =")
print(after_S)
print(f"P(0) = {abs(after_S[0,0])**2:.4f}, P(1) = {abs(after_S[1,0])**2:.4f}")
# Probabilities unchanged, only phase changed

# Verify they are unitary
print(f"\nS unitary? {is_close(S @ S.conj().T, np.eye(2))}")
print(f"T unitary? {is_close(T @ T.conj().T, np.eye(2))}")
```

## Summary Table

| Matrix Type | Property | Quantum Role |
|------------|----------|--------------|
| Hermitian | A = A^dagger | Observables (what you measure) |
| Unitary | U @ U^dagger = I | Quantum gates (operations) |
| Pauli X | Bit flip | Quantum NOT |
| Pauli Y | Bit + phase flip | Rotation around Y axis |
| Pauli Z | Phase flip | Rotation around Z axis |
| Hadamard | Creates superposition | Used in almost every algorithm |
| S, T | Add phase | Fine-grained rotations |
