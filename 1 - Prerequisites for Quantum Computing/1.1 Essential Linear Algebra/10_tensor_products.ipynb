# Chapter 10: Tensor Products — The Quantum Combination

Welcome to the finale of **Section 1.1: Essential Linear Algebra**.

We have mastered single qubits. But a quantum computer with one qubit is just a glorified random number generator.

To do useful computation, we need **many qubits**. We need them to interact. We need them to combine.

How do we describe a system with multiple particles? In classical physics, you just list the state of each particle separately (Particle 1 is here, Particle 2 is there).

In quantum mechanics, we use the **Tensor Product** ($\otimes$). And this mathematical operation gives rise to the weirdest feature of all: **Entanglement**.

---

## 1. Combining Vector Spaces

If System A lives in vector space $V_A$ (dimension $N$),
and System B lives in vector space $V_B$ (dimension $M$),
then the combined system lives in a new space $V_{total} = V_A \otimes V_B$.

The dimension of this new space is $N \times M$.

### Scaling Up
- 1 Qubit: Dimension 2.
- 2 Qubits: Dimension $2 \times 2 = 4$.
- 3 Qubits: Dimension $2 \times 2 \times 2 = 8$.
- $N$ Qubits: Dimension $2^N$.

This is why simulating quantum computers is hard. To write down the state of 300 qubits, you would need more complex numbers than there are atoms in the observable universe.

---

## 2. Calculating the Tensor Product

How do you actually calculate it? We use the **Kronecker Product**.

If $|\psi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$ and $|\phi\rangle = \begin{pmatrix} \gamma \\ \delta \end{pmatrix}$, then:

$$ |\psi\rangle \otimes |\phi\rangle = \begin{pmatrix} \alpha \begin{pmatrix} \gamma \\ \delta \end{pmatrix} \\ \beta \begin{pmatrix} \gamma \\ \delta \end{pmatrix} \end{pmatrix} = \begin{pmatrix} \alpha\gamma \\ \alpha\delta \\ \beta\gamma \\ \beta\delta \end{pmatrix} $$

You take the first element of the first vector, and multiply it by the WHOLE second vector. Then the second element, and so on.

### Example: Basis States

$$ |0\rangle \otimes |0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \otimes \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 1(1) \\ 1(0) \\ 0(1) \\ 0(0) \end{pmatrix} = \begin{pmatrix} 1 \\ 0 \\ 0 \\ 0 \end{pmatrix} = |00\rangle $$

$$ |0\rangle \otimes |1\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix} \otimes \begin{pmatrix} 0 \\ 1 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix} = |01\rangle $$

$$ |1\rangle \otimes |0\rangle = \begin{pmatrix} 0 \\ 0 \\ 1 \\ 0 \end{pmatrix} = |10\rangle $$

$$ |1\rangle \otimes |1\rangle = \begin{pmatrix} 0 \\ 0 \\ 0 \\ 1 \end{pmatrix} = |11\rangle $$

```python
import numpy as np

# Define single qubit states
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

# Calculate Tensor Product using np.kron
ket_00 = np.kron(ket_0, ket_0)
ket_01 = np.kron(ket_0, ket_1)
ket_10 = np.kron(ket_1, ket_0)
ket_11 = np.kron(ket_1, ket_1)

print(f"|00>: {ket_00}")
print(f"|01>: {ket_01}")
print(f"|10>: {ket_10}")
print(f"|11>: {ket_11}")
```

## 3. Entanglement: Breaking the Tensor Product

If we can write a state as $|\psi\rangle \otimes |\phi\rangle$, we say it is a **Product State**. The qubits are independent.

But look at this state (the Bell State):

$$ |\Phi^+\rangle = \frac{1}{\sqrt{2}} (|00\rangle + |11\rangle) = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ 0 \\ 0 \\ 1 \end{pmatrix} $$

Try to find two vectors $\begin{pmatrix} a \\ b \end{pmatrix}$ and $\begin{pmatrix} c \\ d \end{pmatrix}$ such that their product equals this.

$$ \begin{pmatrix} ac \\ ad \\ bc \\ bd \end{pmatrix} = \begin{pmatrix} 1/\sqrt{2} \\ 0 \\ 0 \\ 1/\sqrt{2} \end{pmatrix} $$

1. $ad = 0$. So either $a=0$ or $d=0$.
2. If $a=0$, then $ac=0$. But $ac=1/\sqrt{2}$. Contradiction!
3. If $d=0$, then $bd=0$. But $bd=1/\sqrt{2}$. Contradiction!

**Conclusion:** You CANNOT write this state as a product of two independent qubits. It is impossible.

The qubits are **Entangled**. They have lost their individual identities and merged into a single, shared existence.

---

## 4. Matrix Tensor Products

We also combine gates using tensor products.

If you apply $X$ to qubit 1 and $Z$ to qubit 2, the total operator is $X \otimes Z$.

$$ A \otimes B = \begin{pmatrix} a_{11} B & a_{12} B \\ a_{21} B & a_{22} B \end{pmatrix} $$

This creates a block matrix.

```python
X = np.array([[0, 1], 
              [1, 0]])
Z = np.array([[1, 0], 
              [0, -1]])

# Calculate X tensor Z
XZ = np.kron(X, Z)

print("X tensor Z Matrix:")
print(XZ)

# Let's verify it works as expected
# XZ |00> should be (X|0>) tensor (Z|0>) = |1> tensor |0> = |10>
result = XZ @ ket_00
print(f"(X tensor Z) applied to |00>: {result}")
# Expect: [0, 0, 1, 0] (which is |10>)
```
