# Special Matrices: Hermitian, Unitary, and Projectors

Three classes of matrices dominate quantum mechanics. Understanding their properties is understanding quantum measurement, evolution, and the measurement postulate itself.

## Hermitian Matrices: Observables

A matrix H is **Hermitian** if H† = H (equal to its conjugate transpose):

$$H^\dagger = H \quad \Leftrightarrow \quad H_{ij} = H^*_{ji}$$

**Why Hermitian matters for quantum**:
1. **Real eigenvalues**: λᵢ ∈ ℝ (measurement outcomes must be real!)
2. **Orthogonal eigenvectors**: different eigenvalues give orthogonal eigenstates
3. **Complete basis**: eigenvectors span the entire Hilbert space
4. **Spectral theorem**: H = Σᵢ λᵢ |uᵢ⟩⟨uᵢ| (eigendecomposition)

Examples:
$$\sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} = 1 \cdot |0\rangle\langle 0| - 1 \cdot |1\rangle\langle 1|$$

$$\sigma_x = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} = 1 \cdot |+\rangle\langle +| - 1 \cdot |-\rangle\langle -|$$

where |±⟩ = (|0⟩ ± |1⟩)/√2.

The **measurement postulate** states: upon measuring observable H on state |ψ⟩,
- We get outcome λᵢ with probability P(λᵢ) = |⟨uᵢ|ψ⟩|²
- The state collapses to |uᵢ⟩

This works BECAUSE H is Hermitian!

## Unitary Matrices: Quantum Gates

A matrix U is **unitary** if U†U = I (equivalently, U† = U⁻¹):

$$U^\dagger U = I$$

**Why unitary matters**:
1. **Reversibility**: U⁻¹ = U† always exists (no information lost)
2. **Norm preservation**: ||U|ψ⟩|| = ||ψ⟩|| for all states
3. **Probability conservation**: Σᵢ |⟨i|U|ψ⟩|² = 1 (sum of probabilities = 1)

**All quantum gates are unitary**. This is what makes quantum computing fundamentally different from classical:
- Classical gates (AND, OR) are irreversible—you can't undo AND
- Quantum gates (H, CNOT, T) are always reversible

Eigenvalues of unitary matrices lie on the unit circle: |λᵢ| = 1, so λᵢ = e^{iθᵢ}.

Examples:
$$H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \quad \sigma_x = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$$

Both Hadamard and Pauli-X are unitary (and Hermitian!). Matrices that are both Hermitian AND unitary have eigenvalues ±1.

## Projection Matrices: Measurement Outcomes

A matrix P is a **projector** if P² = P. Geometrically, it projects vectors onto a subspace.

For a pure state |v⟩, the projection operator is:
$$P_v = |v\rangle\langle v|$$

Properties:
- P²ᵥ = |v⟩⟨v|v⟩⟨v| = |v⟩⟨v| = Pᵥ (idempotent) ✓
- Pᵥ is Hermitian: P†ᵥ = Pᵥ
- Eigenvalues are 0 or 1 (either project onto v or annihilate)

**For measurement**: if H = Σᵢ λᵢ Pᵢ where Pᵢ = |uᵢ⟩⟨uᵢ| projects onto eigenstate uᵢ, then:
$$P(\text{measure } \lambda_i) = \text{Tr}(P_i \rho) = |\langle u_i|\psi\rangle|^2$$

Projectors formalize the entire measurement postulate!

## Positive Semidefinite Matrices: Density Matrices

A matrix ρ is **positive semidefinite** if ⟨ψ|ρ|ψ⟩ ≥ 0 for all |ψ⟩. 

**For quantum states**:
- **Pure states**: ρ = |ψ⟩⟨ψ| (Rank 1, entropy 0)
- **Mixed states**: ρ = Σᵢ pᵢ |ψᵢ⟩⟨ψᵢ| (Rank > 1, entropy > 0)

Properties:
- ρ† = ρ (Hermitian)
- Tr(ρ) = 1 (normalized)
- ρ² ≤ ρ (purity check: Tr(ρ²) ≤ 1)
- Eigenvalues ∈ [0,1] (probabilities)

Mixed states represent statistical ensembles—uncertainty from lack of knowledge, not quantum superposition.

## Commuting Observables and Simultaneous Diagonalization

Two observables A,B are **commuting** if [A,B] = AB - BA = 0.

**Spectral theorem for commuting matrices**: If [A,B] = 0, there exists a basis where both A and B are simultaneously diagonal.

Physically: **commuting observables can be measured simultaneously** without disturbing each other.

Example:
$$\sigma_x = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \quad \sigma_z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}$$

$$[\sigma_x, \sigma_z] = \sigma_x \sigma_z - \sigma_z \sigma_x = -2i\sigma_y \neq 0$$

So σₓ and σz don't commute! You CANNOT measure both simultaneously. This is Heisenberg uncertainty at the matrix level.

```python
# Functional code: Special matrices in quantum computing
import numpy as np

# Hermitian matrices (Observables)
sigma_z = np.array([[1, 0], [0, -1]])
sigma_x = np.array([[0, 1], [1, 0]])
H_observable = (sigma_z + sigma_x) / np.sqrt(2)

# Check Hermitian property
is_hermitian_sigz = np.allclose(sigma_z, sigma_z.conj().T)
eigenvalues_sigz = np.linalg.eigvals(sigma_z)  # Should be real

# Unitary matrices (Quantum gates)
H_gate = np.array([[1, 1], [1, -1]]) / np.sqrt(2)  
spec = np.linalg.eigvals(H_gate)  # Should be ±1

# Check unitarity: U†U = I
is_unitary = np.allclose(H_gate.conj().T @ H_gate, np.eye(2))

# Projection matrices: |0⟩⟨0| and |1⟩⟨1|
proj_0 = np.array([[1, 0], [0, 0]])
proj_1 = np.array([[0, 0], [0, 1]])

# Check idempotency: P² = P
idempotent_check = np.allclose(proj_0 @ proj_0, proj_0)

# Projectors as measurement outcomes
psi = (np.array([1, 0]) + np.array([0, 1])) / np.sqrt(2)  # |+⟩
prob_0 = np.real(psi.conj() @ proj_0 @ psi)  # Probability of measuring |0⟩
prob_1 = np.real(psi.conj() @ proj_1 @ psi)  # Probability of measuring |1⟩

# Density matrix: pure state
rho_pure = np.outer(psi, psi.conj())  # |ψ⟩⟨ψ|
purity_pure = np.trace(rho_pure @ rho_pure)  # Tr(ρ²) = 1 for pure

# Density matrix: mixed state
p1, p2 = 0.6, 0.4
rho_mixed = p1 * proj_0 + p2 * proj_1
purity_mixed = np.trace(rho_mixed @ rho_mixed)  # Tr(ρ²) < 1 for mixed

# Commutator: [A,B] = AB - BA
commutator_xz = sigma_x @ sigma_z - sigma_z @ sigma_x
is_commuting = np.allclose(commutator_xz, 0)

# Simultaneous diagonalization (when [A,B] = 0)
# For σz, eigendecomposition:
evals_z, evecs_z = np.linalg.eigh(sigma_z)
```

## Summary: The Three Pillars of Quantum Algebra

✓ **Hermitian observables**: Real eigenvalues (measurement outcomes), orthogonal eigenstates, spectral decomposition  
✓ **Unitary gates**: Always invertible, preserve probabilities, all quantum gates are unitary  
✓ **Projection operators**: Formalize measurement postulate, project onto eigenspaces  
✓ **Density matrices**: Pure states ρ = |ψ⟩⟨ψ|, mixed states from ensembles  
✓ **Commuting observables**: [A,B] = 0 means simultaneous measurement possible  

Together, these three matrix classes encode everything about quantum mechanics.
