# Chapter 07: Special Matrices — The Players

Welcome to **Section 1.1: Essential Linear Algebra**.

Not every matrix is allowed in quantum mechanics. The laws of physics restrict us to very specific types of matrices. Otherwise, probabilities wouldn't add up to 1, or energies would be complex numbers (which makes no sense).

There are two main players you need to know:
1. **Unitary Matrices:** These represent quantum gates (actions).
2. **Hermitian Matrices:** These represent observables (measurements).

If a matrix is not one of these, it essentially doesn't exist in the standard formulation of closed quantum systems.

---

## 1. Unitary Matrices (Gates)

A matrix $U$ is **Unitary** if its conjugate transpose is its inverse.

$$ U^\dagger U = I $$

Or equivalently: $U^\dagger = U^{-1}$.

### Why Unitary?

Unitary transformations **preserve valid quantum states**. Specifically, they preserve the **norm** (length) of the vector.

Proof (sketch):
Let $|\psi'\rangle = U|\psi\rangle$.
The new norm squared is $\langle \psi' | \psi' \rangle$.
Using the dual rule, $\langle \psi'| = \langle \psi | U^\dagger$.
So $\langle \psi' | \psi' \rangle = \langle \psi | U^\dagger U | \psi \rangle$.
If $U^\dagger U = I$, then this equals $\langle \psi | I | \psi \rangle = \langle \psi | \psi \rangle$.

The length is unchanged! If total probability was 100% before, it is 100% after.

Think of Unitary matrices as "Generalized Rotations". They spin the state vector around, but they never stretch or shrink it.

```python
import numpy as np

def is_unitary(M):
    # Check if M_dagger @ M is Identity
    M_dagger = M.conj().T
    product = M_dagger @ M
    return np.allclose(product, np.eye(M.shape[0]))

X = np.array([[0, 1], [1, 0]])
H = (1/np.sqrt(2)) * np.array([[1, 1], [1, -1]])
Random = np.array([[1, 2], [3, 4]])

print(f"Is X unitary? {is_unitary(X)}")
print(f"Is H unitary? {is_unitary(H)}")
print(f"Is Random unitary? {is_unitary(Random)}")
```

---

## 2. Hermitian Matrices (Observables)

A matrix $H$ is **Hermitian** if it is equal to its own conjugate transpose.

$$ H^\dagger = H $$

### Why Hermitian?

Hermitian matrices have a magical property: **all their eigenvalues are real numbers**.

When you measure a physical property (Energy, Spin, Position), the result must be a real number. You cannot measure "3 + 4i Joules" of energy.

Because measurement results are eigenvalues (as we will see in the next notebook), the operators representing physical quantities MUST be Hermitian.

Examples:
- The Hamiltonian (Energy operator)
- Pauli Matrices (Spin operators)
- Position and Momentum operators

```python
def is_hermitian(M):
    return np.allclose(M, M.conj().T)

Z = np.array([[1, 0], [0, -1]])
Y = np.array([[0, -1j], [1j, 0]])

print(f"Is Z hermitian? {is_hermitian(Z)}")
print(f"Is Y hermitian? {is_hermitian(Y)}")
print(f"Is Random hermitian? {is_hermitian(Random)}")
```

## 3. The Pauli Matrices

The most famous matrices in quantum computing are the Pauli Matrices. They are both Unitary AND Hermitian. This is rare and special.

$$ I = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} $$
$$ X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} $$
$$ Y = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix} $$
$$ Z = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} $$

These 4 matrices form a **basis** for all 2x2 complex matrices. Any 2x2 matrix can be written as a sum of these four.

They are the alphabet of single-qubit gates.
