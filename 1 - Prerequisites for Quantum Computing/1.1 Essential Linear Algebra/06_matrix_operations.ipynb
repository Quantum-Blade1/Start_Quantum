# Matrix Operations: The Algebra of Quantum Transformations

Matrix algebra forms the computational backbone of quantum mechanics. Understanding operations on matrices is understanding how quantum states transform, how measurements interact, and how algorithms compose.

## Matrix Addition and Multiplication

**Addition** is element-wise (A + B)ᵢⱼ = Aᵢⱼ + Bᵢⱼ. Physically, this represents superposition of operators:
$$P_0 + P_1 = |0\rangle\langle 0| + |1\rangle\langle 1| = I$$

The sum of projectors onto orthogonal measurement outcomes equals identity.

**Multiplication** A·B applies transformations sequentially: (A·B)|ψ⟩ = A(B|ψ⟩).

**Order matters!** AB ≠ BA in general. This non-commutativity is physically crucial:
- Commuting observables [A,B] = AB - BA = 0 can be simultaneously diagonalized (measured together)
- Non-commuting observables {A,B} cannot be simultaneously measured (Heisenberg uncertainty)

## Adjoint (Conjugate Transpose)

The **adjoint** A† is the complex conjugate transpose. For Hermitian matrices A† = A, which gives:

$$\langle\psi|A|\psi\rangle^* = \langle\psi|A^\dagger|\psi\rangle = \langle\psi|A|\psi\rangle$$

This means expectation values are **always real**—a requirement for observable outcomes.

For unitary matrices U† = U⁻¹, so:
$$U^\dagger U = I \quad \Rightarrow \quad U^\dagger = U^{-1}$$

## Determinant and Invertibility

The **determinant** det(A) tells whether A is invertible. For quantum gates (unitaries), det(U) has magnitude 1:
$$|\det(U)| = 1 \quad \text{(unitary matrices have unit determinant magnitude)}$$

Examples:
- det(H) = -1/2 (magnitude = 0.5, not 1... wait, recalculate!)

Actually det(H) = (1/2)(1 - 1) = 0... No! For H:
$$\det(H) = \frac{1}{2}(1·(-1) - 1·1) = \frac{1}{2}(-2) = -1$$

So |det(H)| = 1 ✓ (Hadamard is unitary, as expected).

The determinant is multiplicative: det(AB) = det(A)det(B), so composing gates preserves unitarity.

## Trace and Eigenvalues

The **trace** (sum of diagonal elements) captures a key property:
$$\text{Tr}(A) = \sum_i A_{ii} = \sum_i \lambda_i$$

where λᵢ are eigenvalues. For unitary gates U, Tr(U) reveals phase information. For measurements, the trace computes expectation values:

$$\langle A \rangle = \text{Tr}(\rho A)$$

where ρ is the **density matrix** (mixed state representation). This formula is the quantum computation formula.

## Inverse and Pseudo-inverse

For invertible A, A⁻¹ satisfies A⁻¹A = I. For unitary U, inverse is simply the adjoint:
$$U^{-1} = U^\dagger$$

This is why quantum gates are reversible with no extra cost—just apply U† to undo U.

**Pseudo-inverse** A⁺ (Moore-Penrose) works even when A is singular. Used in quantum state tomography to reconstruct density matrices from measurement data.

## Matrix Factorizations

**LU Decomposition**: A = LU (lower × upper triangular)
- Used for solving linear systems: A|x⟩ = |b⟩ ⟹ compute |x⟩ via forward/backward substitution

**QR Decomposition**: A = QR (Q unitary, R upper triangular)
- Q preserves norms (useful for Gram-Schmidt orthonormalization)
- Used in quantum error correction to find nearest valid codeword

**Singular Value Decomposition**: A = UΣV† (U,V unitary, Σ diagonal)
- Σ eigenvalues reveal rank of A (important for quantum state preparation)
- Appearing in Schmidt decomposition of entangled states

## Tensor Products of Matrices

For two-qubit gates, combine single-qubit operations via Kronecker product:
$$A \otimes B = \begin{pmatrix} a_{11}B & a_{12}B & \cdots \\ a_{21}B & a_{22}B & \cdots \\ \vdots & \vdots & \ddots \end{pmatrix}$$

Properties:
$$(A \otimes B)(C \otimes D) = (AC) \otimes (BD)$$
$$(A \otimes B)^\dagger = A^\dagger \otimes B^\dagger$$

This means: product states transform independently! 
$$H \otimes H \text{ applies Hadamard to each qubit independently}$$

Physically, non-entangled qubits are described by tensor products; entangled qubits require full 2ⁿ × 2ⁿ matrices.

## Matrix Norms and Error Analysis

The **Frobenius norm** measures gate error:
$$\|A\|_F = \sqrt{\sum_{ij} |A_{ij}|^2}$$

For distinguishing two quantum states with operators A,B:
$$\|A - B\|_F \text{ small} \implies \text{A and B represent nearly identical operations}$$

Quantum algorithm error bounds depend directly on matrix norms of accumulated gate errors.

```python
# Functional code: Matrix operations for quantum gates
import numpy as np

# Define basic gates
H = np.array([[1, 1], [1, -1]]) / np.sqrt(2)
sigma_x = np.array([[0, 1], [1, 0]])
sigma_z = np.array([[1, 0], [0, -1]])

# Matrix addition: sum of projectors
proj_0 = np.outer([1, 0], [1, 0])  # |0⟩⟨0|
proj_1 = np.outer([0, 1], [0, 1])  # |1⟩⟨1|
identity = proj_0 + proj_1  # Should be I

# Matrix multiplication: gate composition
composed = sigma_z @ H  # Apply H then σz
commutator = sigma_x @ sigma_z - sigma_z @ sigma_x  # [σx, σz]

# Adjoint and Hermitian check
H_adjoint = H.conj().T  # H† conjugate transpose
is_hermitian_H = np.allclose(H, H_adjoint)  # Is H = H†? (No, H is not Hermitian)
is_hermitian_sigz = np.allclose(sigma_z, sigma_z.conj().T)  # σz = σz†? (Yes)

# Determinant and  unitarity
det_H = np.linalg.det(H)  # Should have |det| = 1
unitarity_check = np.allclose(H.conj().T @ H, np.eye(2))  # U†U = I?

# Trace
trace_H = np.trace(H)
trace_sigz = np.trace(sigma_z)

# Matrix inverse
H_inverse = np.linalg.inv(H)  # Should equal H† for unitary
H_adjoint_method = H.conj().T

# Singular value decomposition
U, S, Vh = np.linalg.svd(sigma_x)  # σx = U·Σ·V†

# Tensor product of matrices
CNOT_decomposition = np.kron(proj_1, sigma_x) + np.kron(proj_0, np.eye(2))  # Incomplete, illustrative

# Frobenius norm
error_gate = H + 0.01 * np.random.randn(2, 2)
frobenius_norm = np.linalg.norm(H - error_gate, 'fro')
```
