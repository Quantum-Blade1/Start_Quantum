# Chapter 06: Matrix Operations — The Algebra

Welcome to **Section 1.1: Essential Linear Algebra**.

We know what matrices are (operators). Now we look at the rules of how they interact.

In standard algebra ($3 \times 5 = 5 \times 3$), order doesn't matter. In linear algebra ($A \times B \neq B \times A$), **order matters**.

This simple fact — that you cannot swap the order of operations — is the mathematical root of the Heisenberg Uncertainty Principle.

This notebook covers matrix multiplication, transposes, conjugates, and the all-important **Commutator**.

---

## 1. Matrix Multiplication (MatMul)

First, notation.

If we apply gate $B$ first, then gate $A$, the total operation is written as $AB$.

$$ |\psi_{final}\rangle = A(B|\psi_{initial}\rangle) = (AB)|\psi_{initial}\rangle $$

**Crucial:** We read from **right to left**. The matrix closest to the ket hits it first.

### How to Multiply

To get the element at row $i$ and column $j$ of the product $C = AB$, you take the dot product of **Row $i$ of A** and **Column $j$ of B**.

$$ (AB)_{ij} = \sum_k A_{ik} B_{kj} $$

Let's see this in code because doing it by hand is tedious (but you should do it once).

```python
import numpy as np

# Define two matrices
X = np.array([[0, 1], 
              [1, 0]])

Z = np.array([[1, 0], 
              [0, -1]])

# Calculate XZ (apply Z first, then X... wait, no!)
# XZ|psi> means Z is applied first? No.
# (XZ)|psi> = X(Z|psi>). Yes, Z applied first.

XZ = X @ Z

# Calculate ZX (apply X first, then Z)
ZX = Z @ X

print("XZ =")
print(XZ)

print("ZX =")
print(ZX)

# Are they equal?
print(f"Are they equal? {np.array_equal(XZ, ZX)}")
```

---

## 2. No Commutation = Uncertainty

We found that $XZ \neq ZX$. They do not **Commute**.

In physics, if two operators commute, you can know their values simultaneously. If they don't, you can't.

If you measure spin in the Z direction, and then in the X direction, you get a different result than if you measure X then Z. The act of measuring X disturbs the Z value.

We quantify this using the **Commutator** bracket: $[A, B]$.

$$ [A, B] = AB - BA $$

If $[A, B] = 0$, they commute.
If $[A, B] \neq 0$, they don't.

```python
commutator_XZ = XZ - ZX

print("Commutator [X, Z] =")
print(commutator_XZ)

# It is non-zero. Specifically, it is -2iY (roughly).
```

## 3. Transpose ($A^T$)

The transpose flips a matrix over its main diagonal. Rows become columns.

$$ \begin{pmatrix} a & b \\ c & d \end{pmatrix}^T = \begin{pmatrix} a & c \\ b & d \end{pmatrix} $$

---

## 4. Conjugate Transpose ($A^\dagger$)

Also called the **Hermitian conjugate**, **Adjoint**, or **Dagger**.

This is the matrix version of the complex conjugate.

To find $A^\dagger$:
1. Take the transpose ($A^T$).
2. Take the complex conjugate of every element ($^*$).

$$ (A^\dagger)_{ij} = (A_{ji})^* $$

### Why is it fundamental?

In the next notebook, we will see that valid quantum gates (Unitary matrices) are defined by the property $U^\dagger U = I$.

Valid observables (Hermitian matrices) are defined by $H^\dagger = H$.

The dagger is everywhere.

```python
A = np.array([[1, 1-1j], 
              [2, 1j]])

# Calculate Dagger
# Numpy's .conj() gives conjugate. .T gives transpose.
A_dagger = A.conj().T

print("Matrix A:")
print(A)
print("Matrix A dagger:")
print(A_dagger)

# Notice [0,1] was (1-1j). In dagger, [1,0] is (1+1j).
```
