# Vector Spaces: Mathematical Framework for Quantum States

## Formal Definition and Quantum Relevance

A **vector space V** over ℂ (complex field) is a set where:
- Elements can be added: u + v ∈ V (closure)
- Scalars multiply elements: αv ∈ V where α ∈ ℂ

**Quantum significance**: Quantum states ARE vectors in complex vector spaces. This isn't metaphorical—it's the precise mathematical structure.

Ten axioms must hold (associativity, commutativity, identity, inverse, distributivity, scalar multiplication properties). For quantum computing, we work in **ℂⁿ** (n-dimensional complex space) where:

$$v = \begin{pmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{pmatrix}, \quad v_i \in \mathbb{C}$$

## Basis Vectors and Complete Descriptions

Every vector uniquely decomposes in a basis {e₁, e₂, ..., eₙ}:
$$v = c_1 e_1 + c_2 e_2 + \cdots + c_n e_n$$

The coefficients cᵢ are **coordinates** in that basis—they're the quantum amplitudes!

**Standard basis for qubits**: 
$$|0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \quad |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$

Any quantum state: $$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$$

**Critical property**: Different bases reveal different physics. Measuring in computational basis vs X basis vs Y basis gives different outcome distributions—literally different bases show different information.

## Linear Independence and Dimension

Vectors {v₁, ..., vₙ} are **linearly independent** if: c₁v₁ + ... + cₙvₙ = 0 ⟹ c₁ = ... = cₙ = 0

A basis is a maximal linearly independent set. All bases of V have same cardinality (dimension).

**Quantum scaling**: An n-qubit system has dimension **2ⁿ**:
- 1 qubit: ℂ² (dim=2)
- 2 qubits: ℂ⁴ (dim=4)
- 3 qubits: ℂ⁸ (dim=8)
- n qubits: ℂ²ⁿ (exponential growth—core of quantum advantage)

Each basis state |i₁i₂...iₙ⟩ is one dimension. A general state is a superposition of ALL 2ⁿ basis states simultaneously!

## Subspaces, Span, and Tensor Products

A **subspace** W ⊂ V is itself a vector space. Example: The span of {v₁, v₂} is the subspace of all linear combinations.

**Tensor product** V ⊗ W creates larger spaces: 
$$\mathbb{C}^2 \otimes \mathbb{C}^2 = \mathbb{C}^4$$

For multi-qubit systems:
$$\mathcal{H}_n = \mathcal{H}_{qubit1} \otimes \mathcal{H}_{qubit2} \otimes \cdots \otimes \mathcal{H}_{qubitn}$$

Not all states in ℂ⁴ factor as v ⊗ w (those that do are **separable**; those that don't are **entangled**).

## Hilbert Spaces: The Complete Quantum Universe

A **Hilbert space** is a vector space with inner product and completeness. For finite-dimensional quantum systems, this is just ℂⁿ with complex number multiplication.

For mathematical completeness:
- Inner product: ⟨v|w⟩ (measures overlap)
- Completeness: All Cauchy sequences converge (no missing limits)

**All quantum mechanics occurs in Hilbert spaces**. Choosing the right space and basis is half the problem of quantum computing.

## Linear Transformations Between Spaces

A **linear map** T: V → W satisfies:
- T(αu + βv) = αT(u) + βT(v)

Quantum gates are linear transformations on Hilbert space: U|ψ⟩ = |φ⟩

The action is fully determined by where basis vectors go:
- U|0⟩ = ?
- U|1⟩ = ?

This is why quantum gates are 2×2 matrices for single qubits—they're the matrix representation of linear maps on ℂ².
