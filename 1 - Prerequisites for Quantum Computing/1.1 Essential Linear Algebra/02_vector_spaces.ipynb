# Chapter 02: Vector Spaces — The Stage of Quantum Mechanics

Welcome to **Section 1.1: Essential Linear Algebra**.

In the last notebook, we met the **numbers** of quantum mechanics (complex numbers). Now we need a **place** for those numbers to live.

That place is a **Vector Space**.

Imagine a theater. The actors are the quantum states. The script is the Schrödinger equation. But none of it happens without a stage.

In classical mechanics, the "stage" is 3D space. You invoke coordinates $(x, y, z)$.
In quantum mechanics, the "stage" is a sophisticated mathematical object called a **Hilbert Space** (which is just a fancy vector space). It doesn't represent physical space; it represents **state space**.

This notebook will explain exactly what a vector space is, what dimensions really mean, and why a single atom can require an infinite-dimensional stage.

---

## 1. What is a Vector?

Forget whatever you learned about "magnitude and direction" for a moment. That is a physics definition for arrows. In linear algebra, a vector is much more abstract.

**A vector is just an element of a Vector Space.**

That sounds circular, so let's define the space.

A **Vector Space** $V$ is a collection of objects (vectors) that you can:
1. **Add together**: $|v\rangle + |w\rangle$
2. **Multiply by a scalar**: $c|v\rangle$

If you add two vectors, you get another vector in the same space. If you scale a vector, you get another vector in the same space. You can never "fall out" of the space.

### Examples of Vector Spaces
- **Arrows:** The classic example. Add two arrows tip-to-tail, you get a new arrow. Scale an arrow, you get a longer arrow.
- **Polynomials:** $x^2 + 1$ is a vector. $3x - 2$ is a vector. Add them: $x^2 + 3x - 1$. Still a polynomial.
- **Functions:** $f(x) + g(x)$ is a new function.
- **Quantum States:** If $|0\rangle$ is a state and $|1\rangle$ is a state, then $\alpha|0\rangle + \beta|1\rangle$ is a state.

This last point is the definition of **Superposition**. Superposition exists *because* quantum states live in a vector space.

```python
import numpy as np

# Defining vectors in Python
# Numpy arrays are our best representation of vectors

v = np.array([1, 2])
w = np.array([3, -1])

# 1. Vector Addition
sum_v = v + w
# Logic: [1+3, 2+(-1)] = [4, 1]

# 2. Scalar Multiplication
c = 3
scaled_v = c * v
# Logic: [3*1, 3*2] = [3, 6]

print(f"v + w = {sum_v}")
print(f"3 * v = {scaled_v}")

# Linear Combination: The most general operation
# a*v + b*w
a = 2
b = -1
lin_comb = a*v + b*w
print(f"Linear Combination (2v - w): {lin_comb}")
```

## 2. Linear Independence: The "Is it New?" Test

Imagine you are building a Lego set. You have a red brick and a blue brick.
- Can you make a red/blue tower? Yes.
- Can you make a green brick? No.

The green brick is **Linearly Independent** from the red and blue bricks. You cannot build it from what you have.

Mathematically, a set of vectors $\{|v_1\rangle, |v_2\rangle, ...\}$ is linearly independent if no vector in the set can be written as a linear combination of the others.

If $c_1|v_1\rangle + c_2|v_2\rangle = 0$ is only possible when all $c$'s are zero, they are independent.

If you can write $|v_3\rangle = 2|v_1\rangle - |v_2\rangle$, then $|v_3\rangle$ is dependent. It adds nothing new to the space. It is redundant.

## 3. Basis: The Building Blocks

A **Basis** is a minimal set of vectors that can build *any* vector in the space.

Think of primary colors (Red, Green, Blue). With these three, you can mix any color in the rainbow. $\{R, G, B\}$ is a basis for Color Space.

In 2D space (like a sheet of paper):
- Independent vectors: $(1, 0)$ and $(0, 1)$.
- Can you make $(5, 3)$? Yes: $5(1, 0) + 3(0, 1)$.
- Can you make $(100, -20)$? Yes.
- $(1, 0)$ and $(0, 1)$ form a **Basis**.

### The Qubit Basis

For a single qubit, the space is 2-dimensional (over complex numbers). The standard basis is:

$$ |0\rangle = \begin{pmatrix} 1 \\ 0 \end{pmatrix} $$
$$ |1\rangle = \begin{pmatrix} 0 \\ 1 \end{pmatrix} $$

Every valid qubit state is just a mix of these two:
$$ |\psi\rangle = \alpha|0\rangle + \beta|1\rangle $$

This is why we say "a qubit is a superposition of 0 and 1". It means the vector is a linear combination of the basis vectors.

```python
# The Computational Basis in Python
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

# Is this a basis?
# 1. Are they independent? Yes, [1,0] cannot form [0,1].
# 2. Do they span the space? Yes, any [a, b] is a*[1,0] + b*[0,1].

# Let's verify 'spanning'
target_vector = np.array([3, 4])

# Coefficients
alpha = 3
beta = 4

built_vector = alpha * ket_0 + beta * ket_1
print(f"Target: {target_vector}")
print(f"Built: {built_vector}")
print(f"Match? {np.array_equal(target_vector, built_vector)}")
```

## 4. Dimension: The Size of the Stage

The **Dimension** of a vector space is effectively "how many basis vectors do you need?"

- A line is 1D. You need 1 number to specify a point ($x$).
- A plane is 2D. You need 2 numbers ($x, y$).
- The room you are in is 3D ($x, y, z$).
- A qubit is 2D ($\alpha, \beta$).

### The Exponential Explosion

Here is the mind-blowing part of quantum computing.

- 1 Qubit lives in a 2D space ($|0\rangle, |1\rangle$)
- 2 Qubits live in a 4D space ($|00\rangle, |01\rangle, |10\rangle, |11\rangle$)
- 3 Qubits live in an 8D space ($2^3$)
- 10 Qubits live in a 1024D space ($2^{10}$)
- 50 Qubits live in a $1,125,899,906,842,624$D space.

Every time you add **one** qubit, you **double** the dimension of the space.

Classical computers grow linearly (add 1 bit, you have N+1 bits). Quantum computers grow exponentially (add 1 qubit, the state space doubles).

This massive, high-dimensional space is where quantum algorithms perform their magic. They take parallel paths through this enormous landscape.

---

## 5. Spanning a Subspace

Sometimes we only care about a slice of the full space. This is a **Subspace**.

Imagine a 3D room. The floor is a 2D subspace. A line on the floor is a 1D subspace.

In quantum error correction, we often encode information in a specific subspace of a larger system. If the state drifts out of that subspace, we know an error occurred.

For example, in a 3-qubit system (8 dimensions), we might decide that only the states $|000\rangle$ and $|111\rangle$ are valid "code words". These two vectors span a 2D subspace within the larger 8D space. It is a "safe house" inside the larger city.

---

## 6. The "Complex" Twist

One crucial detail: Quantum vector spaces are **Complex Vector Spaces** ($\mathbb{C}^n$).

This means the scalars (the numbers we multiply vectors by) are complex numbers ($a+bi$).

Why? Because nature says so. Experiments show that to describe interference, we need amplitudes that can rotate, not just stretch. Only complex numbers can do that.

So a qubit state vector looks like:
$$ \begin{pmatrix} 1/\sqrt{2} \\ i/\sqrt{2} \end{pmatrix} $$

The coordinates themselves are complex. This makes plotting them hard (interpreting a 2D complex vector requires 4 real dimensions!), but the math works exactly the same as real vectors.

---

## 7. Change of Basis

There is nothing magical about the basis $\{|0\rangle, |1\rangle\}$. We chose it because it's convenient (usually it corresponds to the energy levels of the atom).

But we could choose a different basis. A rotated basis.

$$ |+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle) $$
$$ |-\rangle = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle) $$

The set $\{|+\rangle, |-\rangle\}$ is perfectly valid basis. You can build any qubit state using combinations of $|+\rangle$ and $|-\rangle$. It's just a different perspective. Like looking at a statue from the front vs the side.
