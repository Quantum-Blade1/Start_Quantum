# Chapter 06: Types of Quantum Computing

Welcome to **Section 1.3: General Lecture on Quantum Technology**.

Imagine if, in the 1950s, we had three completely different ways to build a classical computer. One used vacuum tubes, one used gears, and one used water pipes. And nobody knew which one would win.

That is where we are today with Quantum Computing.

There isn't just one way to "compute" with quantum mechanics. There are at least three major paradigms:

1. **Gate-Based Quantum Computing:** The standard model (IBM, Google). Universal. Hard to build.
2. **Quantum Annealing:** Optimization-focused (D-Wave). Easier to build, but limited power.
3. **Topological Quantum Computing:** The dream (Microsoft). Error-free by design, but requires physics that might not exist yet.
4. **Measurement-Based (MBQC):** Compute by measuring (PsiQuantum).

In this notebook, we break them down.

---

## 1. Gate-Based (Universal) QC

**Players:** IBM, Google, Rigetti, IonQ, Honeywell.

**The Logic:** 
This is the model we have been studying. You have qubits. You apply a sequence of logic gates ($H$, $CX$, $T$). You measure the result.

**Pros:**
- **Universal:** Can run ANY quantum algorithm (Shor, Grover, Simulation).
- Well-understood theory.

**Cons:**
- Extremely sensitive to noise. Gates must be perfect.
- Requires complex Error Correction to scale.

---

## 2. Quantum Annealing (Adiabatic)

**Players:** D-Wave Systems.

**The Logic:**
Nature loves low energy. A ball rolls down a hill. An atom falls to its ground state.

Annealing exploits this. You map your problem (e.g., "What is the most efficient delivery route?") to the **energy landscape** of the chip.

1. Start the system in a simple state (superposition of everything).
2. Slowly evolve the system to represent your problem.
3. If you go slow enough (Adiabatic Theorem), the system stays in the lowest energy state.
4. At the end, the configuration of qubits represents the optimal solution.

**Pros:**
- Easier to build. D-Wave has 5000+ qubits today.
- Great for Optimization problems (logistics, finance).

**Cons:**
- **Not Universal.** Cannot run Shor's Algorithm. Cannot run Grover's.
- It is debated whether it provides a true exponential speedup over classical heuristics (Simulated Annealing).

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Visualization of Energy Landscape (Annealing)
# Finding the Global Minimum

def energy_landscape(x, y):
    return (x**2 + y**2) + 10 * np.sin(x) * np.cos(y)

x = np.linspace(-3, 3, 50)
y = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x, y)
Z = energy_landscape(X, Y)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

ax.set_title("Quantum Annealing: Finding the Lowest Valley")
ax.set_xlabel("Qubit Configuration X")
ax.set_ylabel("Qubit Configuration Y")
ax.set_zlabel("Energy (Cost)")

# Highlight the global minimum (approx)
ax.scatter([0], [0], [0], color='red', s=100, label='Global Minimum')
plt.show()
```

## 3. Topological Quantum Computing

**Players:** Microsoft (Station Q).

**The Logic:**
Instead of storing information in a fragile electron, store it in the **global topology** of a system. 

Imagine a string. If you tie a knot in it, the knot is robust. Shaking the string locally doesn't untie the knot. You have to cut the string (global operation).

Microsoft wants to encode qubits in **Anyons** (quasiparticles) that act like knots. We "braid" them in spacetime to perform gates.

**Pros:**
- **Topologically Protected:** Local noise (heat, radiation) cannot destroy the information. The error rate would be essentially zero.
- No need for massive error correction overhead.

**Cons:**
- **It might not work.** We haven't definitively proved we can create and manipulate these Majorana fermions yet. It is high risk, high reward.

---

## 4. Measurement-Based (MBQC)

**Players:** PsiQuantum.

**The Logic:**
In the gate model, you start with $|000\rangle$, apply gates, and measure at the end.

In MBQC (One-Way QC), you do it backwards.
1. Start by creating a massive, highly entangled state (Cluster State) of all qubits.
2. Perform the computation by **measuring** the qubits one by one in specific bases.
3. The measurement of one qubit "teleports" the information to the next, modifying it.

This is popular for Photonic QC because creating entanglement is hard (do it once at the start) but measuring photons is easy.

Most experts believe **Gate-Based** is the long-term winner for universal computing. But Annealers have found a niche in optimization today. Topological is the dark horse that could leapfrog everyone in 2030.
