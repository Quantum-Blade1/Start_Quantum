## Part 1: The Definition (John Preskill)

**The Complexity Class Argument**
Classical computers are in class **P** (Polynomial Time).
Quantum computers are in class **BQP** (Bounded-Error Quantum Polynomial Time).
We believe $BQP \neq P$.
This means there are problems in BQP that are NOT in P.
Quantum Supremacy is the experimental demonstration of this separation.

**The Criteria**
To claim supremacy, you need:
1.  A well-defined computational task.
2.  A programmable quantum computer.
3.  A proof that the task is hard for classical computers (Exponential scaling).
4.  A verification that the quantum computer actually solved it.

## Part 2: Random Circuit Sampling (RCS)

**Why not Factor 21?**
Shor's Algorithm is great, but it requires error correction.
It is too deep for modern hardware.
We need a task that is "shallow but hard".

**The Task**
1.  Take 53 qubits.
2.  Apply a sequence of Random Gates (Hadamard, T, CNOT).
3.  Measure the output bitstring $|x\rangle = |10110...\rangle$.
4.  Repeat 1,000,000 times.

**Why is it hard?**
The output distribution is a "Speckle Pattern".
Some bitstrings are slightly more likely than others due to interference.
To calculate these probabilities classically, you must simulate the full state vector.
State vector size = $2^{53} \times 16$ bytes $\approx 144$ Petabytes.
This swamps the RAM of a Supercomputer.

## Part 4: Cross-Entropy Benchmarking (XEB)

**How do you verify the answer?**
If the classical computer takes 10,000 years, how do you check your work?
Trick: You verify smaller versions (12 qubits, 20 qubits) where you CAN simulate.
You extrapolate the fidelity.

**The Formula**
$$ F_{XEB} = 2^N \langle P(k_i) \rangle - 1 $$
Where $P(k_i)$ is the probability of the observed bitstring.
If the machine is perfect, $F \approx 1$.
If the machine is random noise, $F \approx 0$.
Google measured $F \approx 0.002$ (0.2%).
This sounds low, but it is statistically significant.
It proves they were tracking the quantum interference, not just outputting noise.

```python
import numpy as np
import matplotlib.pyplot as plt

# PART 4 CODE: Porter-Thomas Distribution
# The output probabilities of RCS follow an exponential distribution.

def porter_thomas_demo(num_qubits=10):
    dim = 2**num_qubits
    # Simulate a random state vector
    psi = np.random.randn(dim) + 1j * np.random.randn(dim)
    psi = psi / np.linalg.norm(psi)
    
    probabilities = np.abs(psi)**2
    
    # Scaled probabilities N * p
    scaled_probs = dim * probabilities
    
    # Plot histogram
    plt.figure(figsize=(10, 6))
    plt.hist(scaled_probs, bins=50, density=True, alpha=0.7, color='purple', label='RCS Output')
    
    # Theory: e^-x
    x = np.linspace(0, 10, 100)
    y = np.exp(-x)
    plt.plot(x, y, 'r-', linewidth=3, label='Theory ($e^{-Np}$)')
    
    plt.title("Porter-Thomas Distribution (The Fingerprint of Quantum Chaos)")
    plt.xlabel("Scaled Probability (Np)")
    plt.ylabel("Frequency")
    plt.legend()
    plt.xlim(0, 8)
    plt.grid(True)
    plt.show()

porter_thomas_demo()
```

## Part 8: Complexity Classes (Deep Dive)

**P (Polynomial Time)**
Problems solvable efficiently by a classical computer.
Example: Multiplication, Sorting, Primality Testing.

**BQP (Bounded-Error Quantum Polynomial Time)**
Problems solvable efficiently by a quantum computer.
Example: Factoring (Shor's), Simulation (Hamiltonian), Linear Systems (HHL).

**NP (Nondeterministic Polynomial Time)**
Problems where the *solution* can be verified efficiently.
Example: Sudoku, 3-SAT (Boolean Satisfiability).

**The Hierarchy**
We typically draw it as:
$$ P \subseteq BQP \subseteq PSPACE $$
Notice: We do NOT know if $NP \subseteq BQP$.
Most experts believe **Quantum Computers cannot solve NP-Complete problems efficiently**.
So QCs will probably NOT solve the Traveling Salesman Problem instantly.

## Part 9: The Extended Church-Turing Thesis

**The Original Thesis**
"Anything computable by a human following an algorithm is computable by a Turing Machine."

**The Extended Thesis**
"Any physical process can be simulated efficiently by a probabilistic Turing Machine."

**The Breakdown**
Quantum Supremacy *disproves* the Extended Church-Turing Thesis.
It shows that nature contains a resource (entanglement) that classical Turing Machines cannot simulate efficiently.
This is profound.
It means the universe is fundamentally non-classical.

## Part 11: The Debate: Spoofing

## Part 12: The No-Go Theorems

**No-Cloning Theorem**
You cannot copy an unknown quantum state.
If $|\psi\rangle$ is unknown, there is no Unitary $U$ such that $U(|\psi\rangle|0\rangle) = |\psi\rangle|\psi\rangle$.
This is why we cannot "backup" our quantum data.
This is why error correction is so hard (we can't just majority vote copies).

**No-Deletion Theorem**
You cannot delete a quantum state.
Information is conserved.
This is related to the Black Hole Information Paradox.

**No-Communication Theorem**
Measurement of an entangled particle does not transmit information faster than light.
Quantum mechanics is "Non-Local" but "Signal-Local".
You cannot use entanglement to build a faster-than-light telephone.

## Part 13: Quantum Chaos

**The Signature of Complexity**
Classical Chaos: Sensitive dependence on initial conditions (Butterfly Effect).
Quantum Chaos: The energy levels of the Hamiltonian repel each other.

**Level Spacing Statistics**
Regular System: Poisson Distribution ($P(s) = e^{-s}$).
Chaotic System: Wigner-Dyson Distribution ($P(s) = (\pi s / 2) e^{-\pi s^2 / 4}$).
Random Circuit Sampling essentially creates a generic chaotic Hamiltonian.
This is why the output looks like random speckle.
It is the maximum entropy state.
